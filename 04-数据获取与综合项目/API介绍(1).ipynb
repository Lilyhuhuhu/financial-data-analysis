# API 入门



## 1. 什么是API？


### 1.1 生活中的API类比

想象一下去餐厅用餐的体验：

- **传统方式**：您需要走进厨房，告诉厨师要做什么菜，如何烹饪
- **现代方式**：通过菜单点餐，服务员将您的需求传达给厨房

**API就是数字世界的"菜单和服务员"**，它让不同的软件系统能够相互沟通，而无需了解对方内部的复杂细节。

### 1.2 API的正式定义

API（Application Programming Interface）中文译为"应用程序编程接口"，是一套标准化的规则和工具，让不同的软件能够安全、高效地交换数据和功能。



## 2. 为什么API对商业人士很重要？



### 2.1 现实商业价值

- **成本优化**：无需自建复杂系统，按需使用专业服务

### 2.2 实际应用场景

| 应用领域 | API用途 | 商业价值 |

## 3. 动手体验：从零开始使用API



### 3.1 准备工作：安装必要工具


```python
# 这些代码已经为您准备好，直接运行即可
# 首先安装必要的库
# !pip install requests pandas

# 导入工具包
import requests
print("环境准备完成！")
```


### 3.2 第一个API调用：获取随机用户数据



```python
# 示例1：获取随机用户信息（用于测试和演示）
def get_random_user():
    # 向API发送请求
    response = requests.get('https://randomuser.me/api/')
    
    # 解析返回的数据
    data = response.json()
    
    # 提取有用信息
    user = data['results'][0]
    user_info = {
        '姓名': f"{user['name']['first']} {user['name']['last']}",
        '性别': user['gender'],
        '邮箱': user['email'],
        '城市': user['location']['city'],
        '国家': user['location']['country']
    }
    
    return user_info

# 调用函数并显示结果
user_data = get_random_user()
print("获取到的用户信息：")
for key, value in user_data.items():
    print(f"{key}: {value}")


```


### 3.3 理解API调用过程


请求流程：
您的代码 → 发送请求 → API服务器 → 处理请求 → 返回数据 → 您的代码

关键要素：
1. 端点(Endpoint)：API的"网址"
2. 参数(Parameters)：告诉API您要什么
3. 响应(Response)：API返回的结果


## 4. 常用金融数据API：yfinance 与 Akshare

### 4.1 简介

在金融数据分析中，我们常用的工具有：
1. **yfinance**: 雅虎财经的Python接口，适合获取美股、全球指数等国际数据。
2. **Akshare**: 开源财经数据接口，专为中国市场设计，涵盖A股、期货、基金等。

```python
# 安装必要的库
# !pip install yfinance akshare pandas

import yfinance as yf
import akshare as ak
import pandas as pd
import datetime

print("环境准备完成！")
```

### 4.2 yfinance：全球市场数据

**接口介绍**：
`yfinance` 主要通过 `Ticker` 对象来获取数据。
- `Ticker("代码")`: 初始化股票对象
- `history(period="...")`: 获取历史数据

**调用步骤**：
1. 确定股票代码（如苹果是 AAPL）
2. 创建 Ticker 对象
3. 调用 history 方法获取数据

```python
# 示例：获取苹果公司(AAPL)最近1个月的数据
def get_us_stock_sample():
    # 1. 创建Ticker对象
    apple = yf.Ticker("AAPL")
    
    # 2. 获取历史数据
    df = apple.history(period="1mo")
    
    # 3. 初步清洗：重置索引，只保留收盘价和成交量
    df_clean = df.reset_index()[['Date', 'Close', 'Volume']]
    df_clean['Date'] = df_clean['Date'].dt.date # 只保留日期部分
    
    return df_clean

# 展示数据
us_data = get_us_stock_sample()
print("=== yfinance 样本数据 (AAPL) ===")
print(us_data.head())
```

### 4.3 Akshare：中国市场数据

```python
# 示例：获取贵州茅台(600519)的历史数据
def get_cn_stock_sample():
    # 1. 设定时间范围
    start_date = (datetime.datetime.now() - datetime.timedelta(days=30)).strftime("%Y%m%d")
    end_date = datetime.datetime.now().strftime("%Y%m%d")
    
    # 2. 调用接口 (stock_zh_a_hist)
    try:
        df = ak.stock_zh_a_hist(symbol="600519", period="daily", start_date=start_date, end_date=end_date, adjust="qfq")
        
        # 3. 初步清洗：重命名列，选择需要的字段
        df_clean = df[['日期', '收盘', '成交量']].copy()
        df_clean.columns = ['Date', 'Close', 'Volume']
        
        return df_clean
    except Exception as e:
        print(f"获取失败: {e}")
        return pd.DataFrame()

# 展示数据
cn_data = get_cn_stock_sample()
print("\n=== Akshare 样本数据 (贵州茅台) ===")
print(cn_data.head())
```

## 6. API的类型和使用注意事项


### 6.1 不同类型的API

| API类型 | 特点 | 适用场景 |
|---------|------|----------|
| 开放API | 免费使用，无需注册 | 学习、测试、小项目 |
| 注册API | 需要账号，有免费额度 | 商业分析、个人项目 |
| 付费API | 功能强大，服务质量高 | 企业应用、生产环境 |

### 6.2 使用API的最佳实践

1. **安全第一**：妥善保管API密钥
2. **尊重限制**：遵守调用频率限制
3. **错误处理**：准备好应对API不可用的情况
4. **数据备份**：重要数据及时保存
5. **遵守条款**：尊重服务提供商的使用协议

### 6.3 常见API错误处理



```python
# 安全的API调用方式
def safe_api_call(url):
    try:
        response = requests.get(url, timeout=10)
        response.raise_for_status()  # 检查请求是否成功
        return response.json()
    except requests.exceptions.RequestException as e:
        print(f"API调用失败: {e}")
        return None

# 使用安全调用
user_data = safe_api_call('https://randomuser.me/api/')
if user_data:
    print("API调用成功！")
else:
    print("请检查网络连接或稍后重试")

```
