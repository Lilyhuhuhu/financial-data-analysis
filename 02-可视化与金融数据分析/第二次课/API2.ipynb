## 模块一：API 基础：数字世界的服务窗口 

###  1.1 核心概念：API 是什么？

#### 概念介绍：API 的通俗定义

想象你走进一家提供各种服务的**大型信息服务中心**。

  * **API 就是那个“服务窗口”**。你不用走进复杂的后台去操作（比如，不用知道服务器的内部代码）。
  * 你只需要通过这个窗口，提交你的需求，然后等待它给你结果。

**正式定义：** API (Application Programming Interface)，中文是“应用程序编程接口”。它是一套**固定的规矩**，让不同的电脑程序能安全、高效地互相“聊天”和交换数据。

#### API 的核心价值 (它为什么重要？)

1.  **省事、借力：** 你不用自己去收集或计算数据（比如地图信息、AI 智能），直接向提供这些数据的公司“借力”就行。
2.  **安全保护：** 你拿到的只是你需要的数据，不能看到别人公司的核心秘密（比如数据库密码）。
3.  **标准化：** 所有的请求和回复都遵循一套固定的格式，电脑程序处理起来最方便。

-----

###  1.2 举例：API 的分类与运行模式

#### 应用举例：API 的三种主要类型

| 类型 | 像什么？ | 例子 |
| :--- | :--- | :--- |
| **Web API** (网络 API) | **快递服务** | 通过互联网获取天气、金融汇率或大模型 AI 智能。|
| **库/框架 API** | **工具箱里的螺丝刀** | 编程语言内部自带的功能，比如 Python 的 `print()` 函数。 |
| **操作系统 API** | **电脑的“快捷键”** | 你的程序调用电脑的储存功能、通知功能等。 |

#### API 的运行模式：像寄信一样

1.  **客户端 (Client)：** 你的程序（**寄信人**）。
2.  **API 接口：** 负责收发和传递信息的**邮局**。
3.  **服务器 (Server)：** 提供服务的公司（**收信人**）。

**流程：** 你的程序写好“请求信” → 交给 API 邮局 → 邮局送到服务器 → 服务器处理好数据 → 通过 API 邮局返回“回复包裹”给你。

## 模块二：技术基石：你的请求单和回复格式

###  2.1 概念介绍：如何向 API 提出请求？—————流程三部曲

所有的 API 调用都可以概括为这三个步骤：  

1. 发送请求 (Request): 你的程序通过特定的 URL (地址) 和 HTTP 方法 (例如 GET、POST) 告诉服务器**“我想要什么”**。

2. 接收响应 (Response): 服务器处理请求后，将数据通过网络传回给你的程序。

3. 处理数据 (Process): 你的程序解析接收到的数据 ，提取出你需要的有用信息。

#### 2.1.1 请求地址 (URL)：快递的精确地址

URL 就像你寄信或网购的地址。
`https://api.tianqi.com/weather?city=beijing`

  * `https`：这是安全的传输方式，就像挂号信。
  * `api.tianqi.com/weather`：这是服务窗口的**大门地址**。
  * `?city=beijing`：这是你贴在包裹上的**小纸条**，告诉服务员你的具体要求（查询条件）。

#### 2.1.2 请求动作 (HTTP 方法)：你想要做什么？

API 的动作是固定的，这就是 HTTP 方法，它们告诉服务器你打算做什么。

| 动作 (方法) | 通俗理解 | 目的 (操作) |
| :--- | :--- | :--- |
| **GET** | **我要看 / 查资料** | 从服务器获取数据（最常用）。 |
| **POST** | **我要提交 / 新建记录** | 向服务器发送新数据，比如注册用户。 |
| **PUT/PATCH** | **我要修改** | 更新或修改服务器上已有的信息。 |
| **DELETE** | **我要删除** | 移除一条记录。 |

-----

###  2.2 概念介绍：API 的回复

#### 2.2.1 状态码 (Status Code)：窗口的即时反馈

状态码是服务器对你请求的“盖章”，让你快速知道结果：

| 状态码 | 含义 | 通俗解释 |
| :--- | :--- | :--- |
| **200 OK** | **成功！** | **“搞定！你要的数据在回复包裹里。”** |
| **404 Not Found** | **资源错误** | “地址错了，找不到你要的服务。” |
| **401 Unauthorized** | **权限不足** | “对不起，你没有提供**身份卡（密钥）**，不能查。” |
| **500 Internal Error** | **服务器故障** | “后台厨房出错了，不是你的问题。” |

#### 2.2.2 JSON 数据格式：整洁的清单

API 返回的数据通常是 **JSON** 格式。它就像一个**整理得非常整齐的清单**，让电脑容易看懂。

| JSON 结构 | 像什么？ | 对应 Python |
| :--- | :--- | :--- |
| **对象 (Object)** | **一个档案袋** (里面是各种资料) | **字典 (Dict)** |
| **数组 (Array)** | **一串排队的清单** (里面是重复的项目) | **列表 (List)** |

## 模块三：基础实践：用 Python 迈出第一步
###  举例：获取随机数据
下面例子完美地演示了上面提到的 API 调用三部曲。
1. 示例背景：randomuser.me/api
功能: 这是一个免费且开放的 API 服务，专门用于生成随机的虚假用户数据。

目的: 它经常被程序员用来做测试、演示或填充占位数据，而不是真实的用户数据。

API 地址 (Endpoint): https://randomuser.me/api/。   

```python
import requests # 导入 requests 库，它是 Python 中用于发送 HTTP 请求的常用库
```

```python
# 发出请求
response = requests.get('https://randomuser.me/api/')        
# 解析返回的数据
data = response.json()
print(data)
```

```python
#  处理数据 (Process)    
# 根据 randomuser.me API 的数据结构，用户信息被放在 'results' 键的列表中的第一个元素（索引 0）    
# 提取这个包含用户信息的字典
user = data['results'][0]
user
```

```python
# 从 user 字典中提取我们需要的有用信息，并整理成一个更简洁的字典
user_info = {        
    # 名字和姓氏嵌套在 'name' 字典中，使用 f-string 进行拼接        
    '姓名': f"{user['name']['first']} {user['name']['last']}",        
    # 直接提取性别        
    '性别': user['gender'],        
    # 直接提取邮箱        
    '邮箱': user['email'],        
    # 城市和国家嵌套在 'location' 字典中        
    '城市': user['location']['city'],        
    '国家': user['location']['country']    
}
```

```python
# 显示结果
print("--- 👤 获取到的用户信息 ---")
# 遍历返回的字典，以清晰的格式打印键值对
for key, value in user_info.items():    
    print(f"{key}: {value}")
```

## 模块六：前沿应用：大型语言模型 (LLM) API
### 1. 概念介绍：调用人工智能大脑
#### LLM API 的特殊性
LLM（Large Language Model，大型语言模型，如 Gemini、GPT）API 就像调用了一个**人工智能的超级大脑**。
  * **核心：** 它的价值在于调用了**智能和创造力**，而不仅仅是数据。  
* **Prompt (提示词)：** 这是你发给 AI 的**指令或命令**。Prompt 写得越清楚，AI 返回的结果越好。  
* **角色定义：** 在指令中，你可以让 AI 扮演一个角色（比如“你是专业的记者”），这能让它更准确地完成任务。

### 2. 大模型（LLM）带来的范式转变

大语言模型（如 GPT、通义千问等）的核心在于**预训练**和**指令跟随能力**。它们在海量数据上学习了通用的语言知识，使得文本分析不再局限于任务特定的模型。

| 特性维度 | 传统方法 | 大语言模型 (LLM) |
| :--- | :--- | :--- |
| **知识基础** | **任务专用**：仅从任务训练数据中学习。 | **通用知识**：拥有大规模语料库中积累的背景知识。 |
| **特征提取** | **显性特征工程**：依赖 TF-IDF、词袋等手工设计特征。 | **隐性特征**：通过高维**词嵌入**自动捕捉深层语义和上下文。 |
| **新任务学习** | **必须重新训练**：每个新任务都需要新的标注数据集和完整的模型训练流程。 | **零样本/少样本学习**：通过\*\*Prompt（指令）\*\*即可理解并执行新任务，无需重新训练或大量新标注数据。 |
| **结构化输出** | 困难，通常只能输出简单类别或数值。 | 优秀，可通过 Prompt 严格控制输出格式，轻松生成 **JSON** 或其他复杂结构化数据。 |
| **优势总结** | 速度快，计算资源需求低（一旦部署）。 | 准确率高，泛化能力强，开发效率高，能处理复杂的多维度分析。 |

大模型调用函数编写的思路：  

大模型调用流程抽象成四个核心、通用的步骤，并把其中所有可变的部分（如密钥、地址、模型名称）都设置为函数的输入参数。  

1. 统一的输入（参数化）： 无论调用哪个大模型，都需要知道问什么 (prompt)、去哪里问 (base_url)、拿什么凭证问 (api_key)、用哪个模型问 (model)。将它们全部设为参数，实现了代码与特定厂商的解耦。

2. 构造标准请求： 大部分主流厂商的 API 都遵循了相似的结构（称为 OpenAI/Chat API 兼容规范）：
• Header (请求头)： 使用 Authorization: Bearer <API_KEY> 进行身份验证。<br>
• Payload (请求体)： 包含 model 名称和 messages 列表（通常是 {"role": "user", "content": prompt}）。<br>
4. 发送请求与错误处理： 使用 requests.post() 发送请求，并用 resp.raise_for_status()检查服务器是否返回了错误（例如 404 或 500）。
5. 通用结果解析： 大部分兼容规范的模型，其回复内容都嵌套在 data["choices"][0]["message"]["content"] 这个路径下。函数尝试用这个路径来解析结果，如果解析失败，则抛出错误，提示用户可能需要根据特定模型的返回结构来修改解析逻辑。

```python
import requests
import json
# 确保在运行前安装了 requests 库: pip install requests

def get_completion_universal(
    prompt,        # 【输入参数】用户要向模型提问的问题/内容
    api_key,       # 【输入参数】该服务商（如 DeepSeek, OpenAI）提供的 API 密钥
    base_url,      # 【输入参数】该服务商 API 的访问地址（URL）
    model="default-model", # 【输入参数】要使用的具体模型名称
    temperature=0  # 【输入参数】控制模型随机性和创造性的参数
):
    """
    一个通用的 HTTP API 调用模式，只需传入 API Key 和 Base URL。
    此函数假设模型API遵循标准的 OpenAI/DeepSeek Chat Completions 规范。
    """
    
    # 1. 构造请求头 (Headers)
    headers = {
        # 'Authorization' 字段是认证凭证。'Bearer' 是最常用的 Token 类型。
        "Authorization": f"Bearer {api_key}",  
        # 'Content-Type: application/json' 告诉服务器，请求体是 JSON 格式。
        "Content-Type": "application/json"
    }
    
    # 2. 构造请求体 (Payload/Body)
    # 大部分 Chat API 都要求这个结构：
    payload = {
        "model": model,
        # 'messages' 列表是对话历史，这里只传入一个用户的 prompt
        "messages": [{"role": "user", "content": prompt}],
        "temperature": temperature,
    }
    
    # 3. 核心逻辑：发送 HTTP POST 请求
    resp = requests.post(base_url, headers=headers, json=payload)
    
    # 错误检查：如果 HTTP 状态码是 4xx 或 5xx (客户端/服务器错误)，则抛出异常
    resp.raise_for_status() 
    
    # 4. 解析响应
    data = resp.json() 
    
    # 5. 提取结果
    try:
        # 这是提取模型回复文本最常用的路径
        return data["choices"][0]["message"]["content"]
    except (KeyError, IndexError):
        # 如果模型返回的 JSON 结构与预期的不同，捕获错误并提示用户
        raise ValueError(
            "API 返回结构解析失败，请检查模型文档。"
            # 可以加上 resp.text 帮助调试
        )
```

```python
import os
import requests

# --- 验证配置 ---
API_KEY = os.environ.get("DEEPSEEK_API_KEY", "")
BASE_URL = "https://api.deepseek.com/v1/chat/completions"
MODEL_NAME = "deepseek-chat"

if not API_KEY:
    print("❌ 错误：环境变量 DEEPSEEK_API_KEY 未设置。")
else:
    print("正在发送身份验证请求...")
    try:
        # 1. 构造请求头 (只包含认证信息)
        headers = {
            "Authorization": f"Bearer {API_KEY}",
            "Content-Type": "application/json"
        }
        
        # 2. 构造最小请求体 (只需要 Key 有效，就会成功)
        payload = {
            "model": MODEL_NAME,
            "messages": [{"role": "user", "content": "verify"}],
            "max_tokens": 1  # 最小化消耗
        }

        # 3. 发送 POST 请求并设置超时
        response = requests.post(BASE_URL, headers=headers, json=payload, timeout=5)

        # 4. 核心检查：如果状态码是 2xx (成功)，则通过身份验证
        response.raise_for_status()
        
        # 成功返回
        print("✅ DeepSeek API Key 验证成功！(状态码 200)")

    except requests.exceptions.HTTPError as e:
        # 捕获 HTTP 错误 (如 401 Unauthorized, 403 Forbidden)
        print(f"❌ 身份验证失败：{e}")
        # 打印服务器返回的详细错误信息
        print(f"服务器回复原文: {response.text}")
    
    except requests.exceptions.RequestException as e:
        # 捕获网络连接或超时错误
        print(f"🌐 网络错误或连接超时：{e}")
```

```python
import requests
import json

# === 替换为你自己的 API 密钥 ===
API_KEY = os.environ.get("DASHSCOPE_API_KEY", "")

def call_qwen(model="qwen-turbo", messages=None):
    if messages is None:
        messages = [{"role": "user", "content": "你好"}]
    
    url = "https://dashscope.aliyuncs.com/api/v1/services/aigc/text-generation/generation"  #从官方文档获得
    #url = "https://dashscope.aliyuncs.com/compatible-mode/v1"
    
    headers = {
        "Authorization": f"Bearer {API_KEY}",
        "Content-Type": "application/json"
    }
    
    data = {
        "model": model,
        "input": {
            "messages": messages
        }
    }
    
    response = requests.post(url, headers=headers, json=data)
    return response.json()

# === 测试调用 ===
if __name__ == "__main__":
    result = call_qwen(messages=[{"role": "user", "content": "解释深度学习的三大应用"}])
    
    # print("完整响应:")
    # print(json.dumps(result, indent=2, ensure_ascii=False))
    
    # ✅ 正确提取 text 字段（HTTP API 返回的是 output.text）
    if "output" in result and "text" in result["output"]:
        answer = result["output"]["text"]
        print("\n模型回复:")
        print(answer)
    else:
        print("\n❌ 调用失败或格式异常")
        print("错误信息:", result.get("code", "Unknown error"))
```

## 模块七：总结与未来学习
### 概念介绍：API 的商业价值与认证
#### 7.1.1 API 的商业价值 (为什么大家都在用？)
  * **客户服务：** 通过 API 连接邮件、短信、客服机器人，实现自动化回复和通知。  
* **财务报表：** 通过 API 实时获取股票、汇率、销售数据，自动生成财务报告。  
* **生态系统：** 像 Apple 或 Google 这样的公司，通过开放 API 鼓励其他开发者在它的平台上创造新应用，形成一个强大的生态圈。
#### 7.1.2 进阶认证：OAuth 2.0
  * **通俗解释：** 这是一种更安全的“身份授权”方式。  
* **例子：** 当你登录一个新网站时，选择“使用微信/QQ/Google 登录”。你没有把密码给新网站，只是通过 API 授权了它获取你的公开信息。
### 总结与下一步学习建议
1.  **实战是王道：** 先使用简单、免费的 API（如天气、随机数据）开始练习 GET 请求。
2.  **阅读“菜单”：** 养成阅读 API **官方文档**的习惯，它会告诉你地址、参数和要求。
3.  **选择你的方向：** 根据你的兴趣（金融、地图、AI）选择一个领域的 API 进行深入学习。
4.  **学习 POST：** 熟练掌握 **POST** 方法，这是实现数据写入和与 AI 交互的关键。
