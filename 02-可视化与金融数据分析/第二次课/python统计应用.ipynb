##  1.常见的统计学概念
###  1.1 常见概念

#### 1.1.1 中心趋势
- **均值（Mean）**：数据的平均值，是所有数据点之和除以数据点的数量。
- **中位数（Median）**：数据中间的值，将数据排序后位于中间位置的数值。
- **众数（Mode）**：数据中出现频率最高的值。

#### 1.1.2 离散程度
- **标准差（Standard Deviation）**：衡量数据分布的广度，是每个数据点与均值之间差异的平方和的平均值的平方根。
- **范围（Range）**：数据集中最大值和最小值的差异。

#### 1.1.3 分布形状
- **正态分布（Normal Distribution）**：呈钟形曲线，均值、中位数和众数相等。
- **偏态（Skewness）**：描述数据分布偏离正态分布的程度，可为正偏（右偏）、负偏（左偏）。

#### 1.1.4 统计图表
- **直方图（Histogram）**：通过柱形图显示数据的分布情况。
- **箱线图（Box Plot）**：显示数据的中位数、四分位数和异常值。

---

###  1.2 推论统计学

#### 1.2.1 抽样与总体
- **抽样（Sampling）**：从总体中选择一部分样本进行分析，以代表总体。
- **总体（Population）**：研究对象的全部集合。

#### 1.2.2 估计与假设检验
- **点估计（Point Estimation）**：使用样本数据估计总体参数的值。
- **区间估计（Interval Estimation）**：提供一个估计参数范围的区间。
- **假设检验（Hypothesis Testing）**：通过收集样本数据来进行关于总体的统计推断。

---

###  1.3 相关性与回归分析

#### 1.3.1 相关性
- **相关系数（Correlation Coefficient）**：衡量两个变量之间的线性关系强度和方向。

#### 1.3.2 回归分析
- **简单线性回归**：研究一个自变量和一个因变量之间的关系。
- **多元线性回归**：考虑多个自变量对因变量的影响。

---

###  1.4 统计学在数据分析中的应用
- **数据清理与预处理**：处理缺失值、异常值等。
- **探索性数据分析（EDA）**：通过统计图表和描述统计分析初步了解数据。
- **模型评价**：使用统计方法评估模型的性能。

## 3. 描述统计学

### 3.1 集中趋势 - 均值（Mean）

#### 3.1.1 介绍
均值是统计学中用于衡量一组数据中心位置的指标，它表示数据的平均水平。  
均值是描述数据集集中趋势的基本统计量之一。

#### 3.1.2 原理
均值的计算公式为：

$$
\bar{x} = \frac{1}{n} \sum_{i=1}^{n} x_i
$$

其中：  
- $\bar{x}$ 表示均值  
- $x_i$ 表示第 $i$ 个观测值  
- $n$ 表示观测值的数量  

#### 3.1.3 重点和注意事项
- **重点**：均值提供了数据集中心的一个重要指标，对于初步了解数据整体趋势非常有用。
- **注意事项**：
  - 均值对异常值敏感，极端值可能对均值产生较大影响。
  - 在处理偏态分布的数据时，均值可能不是一个很好的代表。


````python
# 导入需要的库
import numpy as np                  # 用于数值计算（比如求平均值）
import matplotlib.pyplot as plt     # 用于画图（直方图、折线图等）
````

````python
# 准备气温数据（示例：一周每天的温度）
temperature_data = np.array([25, 23, 26, 28, 30, 22, 24])
print(temperature_data)
````

````python
# 计算均值（平均气温）
mean_temperature = np.mean(temperature_data)
print(mean_temperature)
````

````python
# 绘制直方图
plt.hist(
    temperature_data, 
    bins=5,                 # 将数值区间切成 5 段
    alpha=0.7,              # 透明度
    color='green',          # 直方图颜色
    edgecolor='black',      # 边框颜色
    label='Temperature Distribution'  # 图例标签
)

# 在图上画出均值线，并设置标题与坐标标签
plt.axvline(
    mean_temperature,       # 竖线位置：均值
    color='red',
    linestyle='dashed',
    linewidth=2,
    label='Mean'
)

# 添加标题和坐标轴标签
plt.title('Temperature Distribution and Mean for Each Day of the Week')
plt.xlabel('Temperature')
plt.ylabel('Frequency')

# 添加图例
plt.legend()
````

## 2. 集中趋势 - 中位数（Median）

### 2.1 介绍

中位数是一组数据中间值的度量，表示将数据从小到大排列时**位于中间位置的值**。
中位数对于描述数据的中心趋势非常有用，尤其适用于**存在异常值（极端值）**的数据集。

---

### 2.2 原理

中位数的计算方法取决于数据集的大小：

* 若数据集为**奇数个**，则中位数是排序后**位于中间位置**的那个值。
* 若数据集为**偶数个**，则中位数是排序后**中间两个值的平均值**。

---

### 2.3 重点和注意事项

**重点：**

* 中位数**不受极端值影响**，对于偏态分布的数据更加稳健。

**注意事项：**

* 数据集需要进行**从小到大排序**。
* 计算中位数前，通常需要确保数据已经排好序（大部分统计软件会自动完成）。


````python
# 准备考试成绩数据（示例）
exam_scores = np.array([75, 82, 90, 65, 88, 78, 94, 70])

print("考试成绩数据：", exam_scores)
````

````python
# 计算中位数
median_score = np.median(exam_scores)

print("计算得到的中位数为：", median_score)
````

````python
# 绘制箱线图（横向）
plt.boxplot(
    exam_scores,
    vert=False,                     # 设置为横向箱线图
    widths=0.7,                     # 箱体高度
    patch_artist=True,              # 开启填充颜色
    medianprops={'color':'red'},    # 中位数线条颜色
    boxprops={'color':'lightblue'}  # 箱体边框颜色
)

plt.title('Exam Scores Distribution and Median')
plt.xlabel('Exam Scores')
plt.show()

print("箱线图已显示。")
````

````python
# 再次确认中位数输出（便于核对）
print("最终，中位数为：", median_score)
````

## 3. 集中趋势 - 众数（Mode）

### 3.1 介绍

众数是一组数据中**出现频率最高的数值**。
在统计学中，众数是描述数据**典型值（most frequent value）**的重要指标，特别适用于分析数据分布中的“峰值”情况。

---

### 3.2 原理

* **众数定义：**
  众数是数据集中**出现次数最多**的那个值。

* 数据可能：

  * **没有众数**（所有值出现次数相同）
  * **有一个众数**
  * **有多个众数**（称为“多峰分布”）

---

### 3.3 重点和注意事项

**重点：**

* 众数能反映数据中**最典型、最常出现的数值**。
* 对于研究“用户偏好”“市场热点”“金融交易高频价位”等场景非常有用。

**注意事项：**

* 数据集可能没有众数，也可能有多个众数。
* 对于**连续型数据**，很少有完全相同的数值，因此通常需要**分组（binning）**后再判断众数。


````python
# 导入需要的库
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import mode   # 用于计算众数

print("库已成功导入！")
````

````python
# 每日销售额数据（示例）
daily_sales = np.array([200, 150, 200, 250, 300, 200, 150, 250, 200])

print("每日销售额数据为：", daily_sales)
````

````python
# 计算众数（新版 scipy 需要 keepdims=True）
result = mode(daily_sales, keepdims=True)
mode_sales = result.mode[0]

print("计算得到的众数为：", mode_sales)
````

````python
# 绘制销售额直方图
plt.hist(
    daily_sales,
    bins=5,
    alpha=0.7,
    color='orange',
    edgecolor='black',
    label='Sales Distribution'
)

# 在图上画出众数的竖线
plt.axvline(
    mode_sales,
    color='red',
    linestyle='dashed',
    linewidth=2,
    label='Mode'
)

plt.title('Sales Distribution and Mode')
plt.xlabel('Sales')
plt.ylabel('Frequency')
plt.legend()

plt.show()

print("直方图已显示。")
````

## 4. 离散趋势 - 方差（Variance）

### 4.1 介绍

方差是一种用于衡量**数据分散程度**的统计指标。
它描述了每个数据点与数据集平均值之间的**偏离程度**，反映数据的波动大小，是分析数据离散趋势的重要工具。

---

### 4.2 原理

方差的计算方法：

* 对每个数据点，计算其与均值的差值
* 将差值平方（为了避免正负抵消）
* 再对平方后的结果求平均

公式如下：

$$
\text{Variance} = \frac{1}{n} \sum_{i=1}^{n} (x_i - \bar{x})^2
$$

其中：

* $( x_i )$：第 i 个观测值
* $( \bar{x} )$：均值
* $( n )$：样本数量

---

### 4.3 重点和注意事项

**重点：**

* 方差用于描述数据的**总体波动程度**，是金融分析中衡量风险的重要工具（如资产收益率波动）。
* 方差越大，说明数据越分散，波动越大。

**注意事项：**

* 方差对**极端值（异常值）非常敏感**。
* 方差的单位是原始数据单位的**平方**（例如收益率的方差单位为“百分比²”）。
  在解释时需要注意单位含义。

````python
# 数据
exam_scores = np.array([75, 82, 90, 65, 88, 78, 94, 70])

print("考试成绩数据：", exam_scores)
````

````python
# 计算平均值（均值）
mean_score = np.mean(exam_scores)

print("计算得到的均值为：", mean_score)
````

````python
# 直接计算方差
variance = np.var(exam_scores)

print("计算得到的方差为：", variance)
````

````python
plt.hist(
    exam_scores,
    bins=5,
    alpha=0.7,
    color='purple',
    edgecolor='black',
    label='Exam Scores Distribution'
)

plt.title('Exam Scores Distribution and Variance')
plt.xlabel('Exam Scores')
plt.ylabel('Frequency')
plt.legend()

plt.show()

print("直方图已显示。")
````

## 5. 分位数（Quantiles）

### 5.1 介绍

分位数是将数据集按照大小顺序划分为若干**等份**时所对应的数值点。
常见的分位数包括：

* **中位数（50% 分位数）**
* **下四分位数（25% 分位数，Q1）**
* **上四分位数（75% 分位数，Q3）**

分位数能够有效帮助我们理解数据分布的结构，尤其用于了解数据的集中趋势和离散情况。

---

### 5.2 原理

* **中位数（50%）**
  将数据从小到大排序后，位于正中间位置的值，即数据被分成**两等份**。

* **四分位数（25%、50%、75%）**
  将数据集从小到大排序后，将其划分为 **四等份**：

  * **Q1（25%）**：25% 的数据点位于它之下
  * **Q2（50%）**：中位数
  * **Q3（75%）**：75% 的数据点位于它之下

数学公式（适用于排序后数据）：

$$
\text{第 } q \text{ 分位数位置} = \frac{q \cdot (n+1)}{4}
$$

其中：

* $( q )$ 为 1、2、3（对应四分位数 Q1、Q2、Q3）
* $( n )$ 为数据量

> 注：实际使用中，NumPy、Pandas 会使用插值方法自动计算分位数，更为科学与通用。

---

### 5.3 重点和注意事项

**重点：**

* 分位数帮助理解数据的**分布结构**，尤其是数据的中心位置、尾部情况与变异程度。
* 四分位数与箱线图（boxplot）密切相关，可用于识别异常值。

**注意事项：**

* 计算前需将数据**从小到大排序**（软件会自动完成）。
* 中位数是 50% 分位数，四分位数用于描述数据集的**四个区间**。
* 实际数据处理中，分位数算法可能存在不同插值方法，但一般不影响基本理解。

## 计算步骤（分位数）

### **1. 排序数据集**

将考试成绩数据按照从小到大排序。

---

### **2. 计算中位数（Median）**

对于偶数个数据，中位数计算公式为：

$$
\text{中位数} = \frac{X_{\frac{n+1}{2}} + X_{\frac{n+2}{2}}}{2}
$$

---

### **3. 计算四分位数（Quartiles）**

* **下四分位数（Q1）：**

$$
Q_1 = \frac{X_{\frac{n+1}{4}} + X_{\frac{n+2}{4}}}{2}
$$

* **中位数（Q2）：**

（同上中位数的计算方法）

* **上四分位数（Q3）：**

$$
Q_3 = \frac{X_{\frac{3n+1}{4}} + X_{\frac{3n+2}{4}}}{2}
$$

````python
# 考试成绩数据
exam_scores = np.array([75, 82, 90, 65, 88, 78, 94, 70])

print("考试成绩数据：", exam_scores)
````

````python
# 计算中位数（50% 分位数）
median = np.median(exam_scores)

# 计算下四分位数（25%）
q1 = np.percentile(exam_scores, 25)

# 计算上四分位数（75%）
q3 = np.percentile(exam_scores, 75)

print("中位数（Median）为：", median)
print("下四分位数 Q1（25%）为：", q1)
print("上四分位数 Q3（75%）为：", q3)
````

````python
plt.boxplot(
    exam_scores,
    vert=False,                     # 横向箱线图
    widths=0.7,
    patch_artist=True,              # 允许填充颜色
    medianprops={'color':'black'}   # 中位数线的颜色
)

plt.title('Boxplot of Exam Scores with Quartiles')
plt.xlabel('Exam Scores')

plt.show()

print("箱线图已显示。")
````

### 📦 箱线图（Boxplot）解读

#### 🔹 1. 四分位数与中位数

根据前面计算结果：

* **中位数（Median）**：`80.0`
* **第一四分位数 Q1（25%）**：`73.75`
* **第三四分位数 Q3（75%）**：`88.5`

这些数据是箱线图结构的核心基础。

---

#### 🔹 2. 箱线图的组成及含义

##### **① 中位数（箱体中间的粗线）**

* 表示一半的学生成绩高于该值，一半低于该值。
* 当前中位数为 **80**，说明总体成绩偏高。

##### **② 箱体下边界（Q1）**

* 第一四分位数为 **73.75**，意味着 **25% 的学生分数低于 73.75**。

##### **③ 箱体上边界（Q3）**

* 第三四分位数为 **88.5**，意味着 **75% 的学生分数低于 88.5**。

##### **④ 四分位距（IQR）**

* 计算方式：

  ```text
  IQR = Q3 - Q1 = 88.5 - 73.75 = 14.75
  ```
* 表示“中间 50%”学生成绩的分布宽度。
* IQR 较小说明成绩差异不大，集中在比较窄的区间内。

##### **⑤ 触须（Whiskers）**

* 从箱子延伸出去的线表示“非异常值”的最大值和最小值。
* 展示整体成绩分布的范围。

##### **⑥ 异常值（若有）**

* 图中没有独立小圆点，说明没有显著偏离整体的异常分数。

---

#### 🔹 3. 综合解释

***成绩整体优秀**：中位数 80 分表明多数学生成绩较好。
***分布较集中**：IQR 只有 14.75，说明学生成绩差距不大。
***高分占比大**：75% 的学生分数低于 88.5，成绩结构偏高。
***无异常值**：没有过高或过低的极端值，说明成绩稳定。

````python
# 另外一组实验

# -----------------------------
# 数据（含异常值）
# -----------------------------
scores = np.array([70, 72, 75, 78, 80, 82, 85, 88, 90, 92, 30, 100])

# -----------------------------
# 计算统计量
# -----------------------------
median = np.median(scores)
q1 = np.percentile(scores, 25)
q3 = np.percentile(scores, 75)
iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

# 非异常值范围内的最大最小值（触须）
lower_whisker = np.min(scores[scores >= lower_bound])
upper_whisker = np.max(scores[scores <= upper_bound])

# 异常值
outliers = scores[(scores < lower_bound) | (scores > upper_bound)]

# -----------------------------
# 画箱线图
# -----------------------------
plt.figure(figsize=(10,5))

plt.boxplot(
    scores,
    vert=False,
    patch_artist=True,
    medianprops={"color": "black"},
    flierprops={"marker": "o", "markerfacecolor": "red", "markersize": 8}
)

plt.title("Boxplot with Automatic Explanation")
plt.xlabel("Scores")

plt.show()

# -----------------------------
# 自动打印解释说明
# -----------------------------
print("📌 箱线图自动解读：\n")
print(f"• 中位数（Median）为：{median}")
print(f"• 下四分位数 Q1（25%）：{q1}")
print(f"• 上四分位数 Q3（75%）：{q3}")
print(f"• 四分位距 IQR = Q3 - Q1 = {iqr}")

print("\n📌 触须（Whiskers）解释：")
print(f"• 下触须（非异常值的最小值）为：{lower_whisker}")
print(f"• 上触须（非异常值的最大值）为：{upper_whisker}")
print(f"  （注意：触须 ≠ 全体最大/最小值，而是正常范围内的最大/最小值）")

print("\n📌 异常值（Outliers）：")
if len(outliers) == 0:
    print("• 无异常值")
else:
    print(f"• 异常值共有 {len(outliers)} 个：{outliers}")

print("\n📌 综合解释：")
print(
    f"这幅箱线图显示，大部分分数集中在 {q1:.2f} 到 {q3:.2f} 区间，中位数是 {median} 分。"
    f"触须从 {lower_whisker} 延伸到 {upper_whisker}，表示非异常值的范围；"
    f"而像 {list(outliers)} 这样的分数因超出 1.5×IQR 被识别为异常值并以红点显示。"
)
````

## 6、数据基本特征描述（Descriptive Statistics）

### 6.1 描述性统计分析

#### 6.1.1 介绍

描述性统计分析是统计学中用于**理解、总结和概括数据集**的基础方法。
它通过一系列指标展示数据的核心特征，包括分布形态、集中趋势、离散程度和偏态等方面。

这些指标能够帮助建立对数据整体结构的认识，是进行任何进一步统计分析或建模的前提。

---

#### 6.1.2 原理

描述性统计分析的主要指标包括：

* **均值（Mean）**：
  衡量数据的中心趋势，表示数据的平均水平。

* **中位数（Median）**：
  数据的中间位置值，对极端值具有较强的稳健性。

* **标准差（Standard Deviation）**：
  衡量数据离散程度，反映数据围绕均值的波动大小。

* **偏态系数（Skewness）**：
  衡量数据分布是否对称。
  正偏态表示右侧尾部较长，负偏态表示左侧尾部较长。

* **四分位数（Quartiles）**：
  将数据集划分为四等份，包括

  * 下四分位数（Q1，25%）
  * 中位数（Q2，50%）
  * 上四分位数（Q3，75%）

这些指标共同构成了对数据集**中心、分散、形态**的全面描述。

---

#### 6.1.3 重点和注意事项

**重点：**

* 描述性统计能够迅速提供数据的**整体特征与结构**。
* 结合多个指标使用，可以更全面地理解数据，而不仅依赖单一指标。

**注意事项：**

* 不同指标关注数据的不同方面，需要综合使用。
* **异常值（outliers）**可能显著影响均值、标准差等指标，分析时需谨慎处理。
* 对偏态或重尾分布的数据，应重点关注分位数、四分位距等稳健指标。

## 7、多元数据的数据特征

### 7.1 协方差与相关系数的概念

---

### 7.1.1 介绍

协方差（Covariance）和相关系数（Correlation Coefficient）是衡量**两个变量之间线性关系强弱**的重要统计指标。
它们帮助我们理解两个变量是否：

* 同向变化（正相关）
* 反向变化（负相关）
* 无明显线性关系（不相关）

---

### 7.1.2 原理

#### **（1）协方差 Covariance**

协方差用于衡量两个变量在各自均值基础上的**同步偏移程度**：

* 协方差 > 0：两个变量倾向于同时增加或减少 → **正相关**
* 协方差 < 0：一个增加一个减少 → **负相关**
* 协方差 = 0：没有线性关系（但不代表没有非线性关系）

协方差公式为：

$$
\text{Cov}(X, Y) = \frac{\sum_{i=1}^{n} (X_i - \bar{X})(Y_i - \bar{Y})}{n}
$$

---

#### **（2）相关系数 Correlation Coefficient**

相关系数是**标准化后的协方差**，用于消除量纲影响，使结果更容易解释。

* 取值范围：
  $$
  -1 \le \rho_{XY} \le 1
  $$

* 解释：

  * ρ = 1：完全正相关
  * ρ = -1：完全负相关
  * ρ = 0：无线性相关

相关系数的公式为：

$$
\rho_{XY} = \frac{\text{Cov}(X, Y)}{\sigma_X \sigma_Y}
$$

其中：

* $( \sigma_X )$：X 的标准差
* $( \sigma_Y )$：Y 的标准差

---

### 7.1.3 重点和注意事项

**重点：**

* 协方差与相关系数是理解两个变量关系的关键指标。
* 协方差关注“方向”，相关系数关注“强度”。

**注意事项：**

* 协方差的单位是两个变量单位的乘积，例如：

  * X：元（收入）
  * Y：岁（年龄）
  * Cov(X,Y) 单位 = 元·岁
    → 不便于直接比较变量之间的关系。
* 相关系数是无量纲的，更方便比较不同变量之间的关系。
* 相关系数越接近 ±1，关系越强；越接近 0，关系越弱。

### ✅ 一、什么是 PCA？

**主成分分析（PCA）就是：
把多个高度相关的变量压缩成少数几个“互不相关”的综合指标（主成分），同时尽量保留原始数据的信息。**

---

### ✅ 二、用成绩案例讲清 PCA

原始数据有 3 个变量：

* 数学（Math）
* 物理（Physics）
* 化学（Chemistry）

这三门课大概率是相关的：
**理科好的学生通常三门都高，理科弱的学生三门都低。**

PCA 会做三件事：

---

#### 👉（1）找出方差最大的方向 —— 第一主成分（PC1）

第一主成分 = 原始三个科目的**最佳“加权平均”**，权重由算法决定。

它的作用：
**尽可能解释“总体成绩差异”**。

课堂语言：

> PC1 就是“综合理科能力指数”。

---

#### 👉（2）找出剩余信息的第二大方向 —— 第二主成分（PC2）

PC2 解释 PC1 解释不了的差异：

比如：

* 有的学生数学和物理高，但化学一般
* 有的学生化学特别好，但数学弱

它反映的是：
**偏科程度 / 非均衡能力**。

课堂语言：

> PC2 = “偏科指数”。

---

#### 👉（3）最后的主成分（PC3）

解释非常少的一部分变化，很多时候可以忽略，用于“噪音”。

---

### ✅ 三、为什么 PCA 会得到这些结论？

PCA 的数学逻辑非常简单：

#### ✔ 第一步：把数据当成 3 维空间点（Math, Phys, Chem）

例如每个学生是一个 3D 点。

#### ✔ 第二步：找出数据分布最“拉长”的方向

这个方向就是 PC1（方差最大 → 信息最多）

#### ✔ 第三步：找出与 PC1 垂直、但仍能解释次多方差的方向

这是 PC2

#### ✔ PC1 ⟂ PC2 ⟂ PC3

所有主成分之间都互相**不相关（orthogonal）**。

---

### ✅ 四、金融领域的典型应用

#### ✔（1）PCA 就是因子模型的“自动版”

很多股票一起涨跌 → 有共同因子（市场因子、行业因子等）

PCA 的含义就是：

> “让电脑自动找出最能解释这些资产共同运动的因子”

比如股票收益率矩阵（100 支股票 × 100 天收益率）做 PCA 时：

* PC1 ≈ 市场因子
* PC2 ≈ 行业结构
* PC3 ≈ 风险风格变化

---

#### ✔（2）为什么 PCA 有用？

* 可以把**很多变量 → 少数几个核心因子**
* 因子之间**不相关**，利于建模
* 最大限度**保留信息**
* 能用于降维、可视化、风险分析

---

# 8、假设检验（Hypothesis Testing）

## 8.1 零假设与拒绝零假设

### 8.1.1 介绍

假设检验是统计学中通过**样本数据**对**总体参数**进行判断的一种方法。
核心思想是：

> 先提出一个默认成立的假设（零假设 H0），
> 再用样本数据检验是否有足够证据推翻它（拒绝零假设 H1）。

零假设（H0）与拒绝零假设（H1）构成假设检验的基本框架。

---

### 8.1.2 原理

* **零假设（H0）**
  表示“没有效应、没有差异、没有关联”。
  它是一种“默认状态”，除非证据非常强，否则不轻易否定。

  数学表达为：

  $$
  H_0: \mu = \mu_0 \quad \text{或} \quad H_0:\text{两组数据没有差异}
  $$

---

* **拒绝零假设（H1）**
  又称备择假设，表示“存在效应、存在差异、存在关联”。
  它是研究者真正关心的推断方向。

  表达为：

  $$
  H_1: \mu \neq \mu_0 \quad \text{或} \quad H_1:\text{两组数据存在差异}
  $$

---

### 8.1.3 重点和注意事项

**重点：**

* 假设检验的本质是判断：
  **样本数据是否提供足够证据可以拒绝 H0？**
* 结论永远是：

  * “拒绝 H0”
  * 或 “不能拒绝 H0”（而不是“接受 H0”）

---

**注意事项：**

* **类型 I 错误（α 错误）**
  错误地拒绝了真实的零假设。
  （“把对的当成错的”）

* **类型 II 错误（β 错误）**
  错误地未拒绝虚假的零假设。
  （“把错的当成对的”）

* **显著性水平（α）**
  控制拒绝 H0 的“严格程度”，常见值：0.05、0.01
  例如：
  α = 0.05 表示有 5% 的风险犯 I 类错误。

### 案例：两个购物车页面转化率是否有显著差异？

某电商公司正在测试两种不同设计的购物车结算页面，想知道它们的**转化率是否存在显著差异**。
设定两个页面的转化率分别为：

* A 网站转化率：( p_A )
* B 网站转化率：( p_B )

#### **数据：**

* **A 网站**

  * 购物车结算成功用户数：80
  * 总用户数：1000

* **B 网站**

  * 购物车结算成功用户数：120
  * 总用户数：1000

---

#### 假设构建

* **零假设（H0）**：两个网站的转化率相等
  $$
  H_0: p_A = p_B
  $$

* **备择假设（H1）**：两个网站的转化率不相等
  $$
  H_1: p_A \neq p_B
  $$

该检验属于 **两独立样本比例差异的 Z 检验（Two-proportion Z-test）**。

---

#### 计算步骤

##### **1. 计算样本转化率**

$$
\hat{p}_A = \frac{\text{A 网站成功人数}}{\text{A 网站总人数}}
$$

$$
\hat{p}_B = \frac{\text{B 网站成功人数}}{\text{B 网站总人数}}
$$

---

##### **2. 计算整体合并转化率**

$$
\hat{p} = \frac{\text{A 成功人数} + \text{B 成功人数}}{n_A + n_B}
$$

再计算标准误差：

$$
SE = \sqrt{\hat{p}(1-\hat{p})\left(\frac{1}{n_A} + \frac{1}{n_B}\right)}
$$

其中

* $( n_A )$：A 网站样本量
* $( n_B )$：B 网站样本量

---

##### **3. 计算 Z 统计量**

$$
Z = \frac{\hat{p}_A - \hat{p}_B}{SE}
$$

---

##### **4. 进行假设检验**

* 选择显著性水平：通常为 **0.05**
* 查标准正态分布临界值：

  * 双尾检验临界值为 **±1.96**

**决策规则：**

$$
|Z| > 1.96 \quad \Rightarrow \quad \text{拒绝零假设}
$$


````python
import numpy as np
import scipy.stats as stats

print("库导入完成！")
````

````python
# 数据
success_A, total_A = 80, 1000   # A 网站：成功 80 / 总数 1000
success_B, total_B = 120, 1000  # B 网站：成功 120 / 总数 1000

print("A网站成功数/总数：", success_A, "/", total_A)
print("B网站成功数/总数：", success_B, "/", total_B)
````

````python
# 转化率计算
p_A = success_A / total_A
p_B = success_B / total_B

print("A网站转化率 p_A：", p_A)
print("B网站转化率 p_B：", p_B)
````

````python
# 汇总转化率（合并样本）
p_combined = (success_A + success_B) / (total_A + total_B)
print("汇总转化率 p：", p_combined)
````

````python
# 计算标准误差 SE
SE = np.sqrt(p_combined * (1 - p_combined) * (1 / total_A + 1 / total_B))

print("标准误差 SE：", SE)
````

````python
# Z统计量
Z = (p_A - p_B) / SE
print("Z统计量：", Z)
````

````python
# P 值（双侧）
p_value = 2 * (1 - stats.norm.cdf(np.abs(Z)))
print("P值：", p_value)
````

````python
alpha = 0.05
if p_value < alpha:
    print("拒绝零假设：两个网站转化率存在显著差异。")
else:
    print("未拒绝零假设：没有足够证据表明两个网站转化率不同。")
````

````python
# 用内置函数进行检验
from statsmodels.stats.proportion import proportions_ztest

# 数据
success_counts = np.array([80, 120])   # 成功人数
sample_sizes = np.array([1000, 1000])  # 总人数

# 双侧检验
z_stat, p_value = proportions_ztest(count=success_counts, nobs=sample_sizes)

print("Z统计量:", z_stat)
print("P值:", p_value)

alpha = 0.05
if p_value < alpha:
    print("拒绝零假设：两个网站转化率存在显著差异。")
else:
    print("未拒绝零假设：没有证据显示转化率不同。")
````

## 8.2 基本原理 - 显著性水平的选择（Significance Level）

### 8.2.1 介绍

显著性水平是假设检验中的核心参数，用于决定我们在什么情况下**拒绝零假设（H0）**。
通常记为 **( \alpha )**，它代表在零假设为真时，我们愿意接受 **错误拒绝它（第一类错误）** 的最大概率。

---

### 8.2.2 原理

* **显著性水平 ( \alpha )**：
  表示在零假设真实成立的情况下，我们仍然拒绝零假设的概率。

常用的显著性水平包括：

* **0.05（最常用）** → 容忍 5% 的第一类错误概率
* **0.01** → 更严格，容忍 1% 的第一类错误概率

显著性水平用于与检验统计量（如 Z 值、t 值）对应的 **P 值** 或 **临界值** 进行比较，决定是否拒绝零假设。

---

### 8.2.3 重点和注意事项

**重点：**

* 显著性水平影响最终的假设检验结论。
* 常用 **0.05** 作为标准，即“有 95% 的把握认为结果不是偶然”。

**注意事项：**

* 显著性水平越小，越严格，**第一类错误概率更小**，但可能导致 **第二类错误概率更大**（即应拒绝 H0 却没有拒绝）。
* 显著性水平需根据：

  * 研究领域规范（如医学常用 0.01）
  * 实际业务风险
  * 决策成本
    进行合理选择。

#### 案例：新药物治愈率的显著性检验（显著性水平 $\alpha = 0.05$）

某医药公司研发了一种新药物，希望证明该药物的治疗效果**显著提高**。我们设定显著性水平为 $0.05$，用假设检验来判断新药物是否显著提高疾病的治愈率。

**数据：**

* 新药物治愈的患者数：$90$
* 总的治疗患者数：$1000$

---

##### 零假设与备择假设

* **零假设（$H_0$）**：新药物的治愈率没有显著提高
  $$H_0: p \leq p_0$$

* **备择假设（$H_1$）**：新药物的治愈率显著提高
  $$H_1: p > p_0$$

其中：

* $p$ 表示新药物的真实治愈率
* $p_0$ 表示旧药物的治愈率（对照基准）

---

##### 计算步骤

1. **计算样本治愈率：**

   $$\hat{p} = \frac{\text{新药物治愈的患者数}}{\text{总的治疗患者数}}$$

2. **计算标准误差（Standard Error）：**

   $$SE = \sqrt{\frac{\hat{p}\left(1 - \hat{p}\right)}{n}}$$

   其中，$n$ 为总的治疗患者数。

3. **计算 $Z$ 统计量：**

   $$Z = \frac{\hat{p} - p_0}{SE}$$

4. **假设检验：**

   * 显著性水平选择为 $0.05$。
   * 查标准正态分布表得到临界值 $z_{0.95}$。
   * 若 $Z$ 大于临界值，则**拒绝零假设** $H_0$，认为新药物治愈率**显著提高**；否则，**不拒绝零假设**。

````python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt

print("库导入完成！")
````

````python
# 新药物治愈数据
cured_patients = 90      # 新药物治愈人数
total_patients = 1000    # 总治疗人数

print("治愈人数：", cured_patients)
print("总人数：", total_patients)
````

````python
# 样本治愈率 p̂
p_hat = cured_patients / total_patients

print("样本治愈率 p̂ =", p_hat)
````

````python
# 显著性水平
alpha = 0.05
print("显著性水平 α =", alpha)
````

````python
# 标准误差 SE
SE = np.sqrt(p_hat * (1 - p_hat) / total_patients)

print("标准误差 SE =", SE)
````

````python
# Z统计量计算
p0 = 0.2      # 旧药物治愈率
Z = (p_hat - p0) / SE

print("Z 统计量 =", Z)
````

````python
# P值（右尾检验）
p_value = 1 - stats.norm.cdf(Z)

print("P 值 =", p_value)
````

````python
# 基于显著性水平的判断
if p_value < alpha:
    print("结果：拒绝零假设，新药物治愈率显著提高。")
else:
    print("结果：未拒绝零假设，没有足够证据表明新药物治愈率提高。")
````

````python
# 直接使用内置函数计算

from statsmodels.stats.proportion import proportions_ztest

cured_patients = 90
total_patients = 1000
p0 = 0.2   # 旧药治愈率假设值

z_stat, p_value = proportions_ztest(
    count=cured_patients,
    nobs=total_patients,
    value=p0,
    alternative='larger'   # H1: p > p0
)

print("Z统计量:", z_stat)
print("P值:", p_value)

# 基于显著性水平的判断 
# 显著性水平 
alpha = 0.05
if p_value < alpha:     
    print("结果：拒绝零假设，新药物治愈率显著提高。") 
else:     
    print("结果：未拒绝零假设，没有足够证据表明新药物治愈率提高。")
````
