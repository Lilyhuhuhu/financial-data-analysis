```python
# ================= 宏观数据原始导入  =================

macro = {}  # 用 dict 统一管理

macro["cpi"]  = ak.macro_china_cpi()   # CPI
macro["ppi"]  = ak.macro_china_ppi()   # PPI
macro["pmi"]  = ak.macro_china_pmi()   # PMI

for name, df in macro.items():
    print(f"\n=== {name.upper()} 前5行 & 列信息 ===")
    print(df.head())
    print("列名：", df.columns.tolist())
    print(" dtypes:\n", df.dtypes)
```

```python
# ============== 只保留当月数据 ==============

def clean_month_cn(col):          # 函数：中文月份 → 日期
    return pd.to_datetime(col.str.replace("年", "-").str.replace("月份", ""), format="%Y-%m")

start_str, end_str = "2020-01", "2023-12"   # 要的时间段

# ① CPI
cpi = macro["cpi"][["月份", "全国-当月"]].copy()   # 只留两列
cpi["月份"] = clean_month_cn(cpi["月份"])
cpi = cpi[cpi["月份"].between(start_str, end_str)]
cpi.rename(columns={"全国-当月": "CPI"}, inplace=True)   # 列名简洁化

# ② PPI
ppi = macro["ppi"][["月份", "当月"]].copy()
ppi["月份"] = clean_month_cn(ppi["月份"])
ppi = ppi[ppi["月份"].between(start_str, end_str)]
ppi.rename(columns={"当月": "PPI"}, inplace=True)

# ③ PMI
pmi = macro["pmi"][["月份", "制造业-指数"]].copy()
pmi["月份"] = clean_month_cn(pmi["月份"])
pmi = pmi[pmi["月份"].between(start_str, end_str)]
pmi.rename(columns={"制造业-指数": "PMI"}, inplace=True)

print(">>> 清洗后样本（头5行）")
print(cpi.head(), "\n")
print(ppi.head(), "\n")
print(pmi.head())
```
