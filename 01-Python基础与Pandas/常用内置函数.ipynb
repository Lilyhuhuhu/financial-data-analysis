### 数学相关

| 函数                        | 功能                                                         | 例子              | 运行结果    |
| --------------------------- | ------------------------------------------------------------ | ----------------- | ----------- |
| abs(a)                      | 对a取绝对值                                                  | abs(-1)           | 1           |

# 2. 读取数据（csv / excel）

数据分析最常见的第一步就是读取外部数据。Pandas 提供了最常用的两个函数：

函数	作用
pd.read_csv(filepath, encoding=None)	读取 .csv 文件
pd.read_excel(filepath, encoding=None)	读取 .xls / .xlsx 文件
1. 为什么需要 encoding？

因为不同来源的数据编码格式不同。例如：

UTF-8：国际通用编码（网站数据一般是 UTF-8）

GBK：中文 Windows 很多旧系统使用的编码

如果编码不匹配，会出现 乱码 或 读取错误。

open(file, mode='r', encoding=None)
- file 文件路径
- mode 操作方式们，最常用的是r和a+。r读取， a+是追加写入
- encoding 编码方式 ，常见的文件编码方式主要是utf-8和gbk

读取返回io对象

io对象有read()方法

###《关于路径——绝对路径 & 相对路径 》
> 路径是什么？
> 就是电脑世界里的“地址”。
> 我们把文件放在哪、去哪找它，都靠这串“地址”来定位。

而“绝对路径”和“相对路径”，其实就像两种不同的——
**找对象的方式**。😉

---

#### 什么是绝对路径？（Absolute Path）

##### ✔ 定义

**绝对路径 = 从电脑系统的最顶层，一路写到文件所在位置的完整路径。**

它不依赖当前的位置，不管你现在在哪个文件夹，都能找到它。

##### ✔ 生活类比（甜甜的😋）

就像说：

> “你给我你的家庭住址，我从地球上任何地方都能找到你。”

非常官方、非常精确。

##### ✔ 示例

##### **Windows：**

```
C:\Users\Lenovo\Desktop\data\iris.csv
```

##### **Mac / Linux：**

```
/Users/huliqin/Desktop/data/iris.csv
```

##### ✔ 特点

* 从根目录（C: 或 /）开始
* 写得很长
* 永远不会迷路
* 就算你换个位置运行代码，也能找到这个文件

##### ✔ Pandas 读取方式

```python
df = pd.read_csv('C:/Users/Lenovo/Desktop/data/iris.csv')
```

---

#### 什么是相对路径？（Relative Path）

##### ✔ 定义

**相对路径 = 相对于你当前工作目录（Current Working Directory）的位置来寻找文件。**

就像告诉别人：

> “你现在在哪里？
> 我就在你旁边的房间。”

比绝对路径短得多，也更灵活。

##### ✔ 生活类比

你和朋友在商场里。
朋友问：“你在哪？”
你说：

> “我就在你前面两家店，过来就能看到我！”

这就是相对位置，不需要从“北京市 → 朝阳区 → xx 商场 → x 层”这样讲一遍。

##### 示例（假设你当前工作目录在 `Desktop/Project/`）

| 相对路径写法            | 含义               |
| ----------------- | ---------------- |
| `data/iris.csv`   | 在当前目录下的 data 文件夹 |
| `./data/iris.csv` | 也是当前目录（“./”可省略）  |
| `../file.csv`     | 返回上一层目录          |
| `../../abc.csv`   | 往上两层             |

##### ✔ Pandas 读取方式

```python
df = pd.read_csv('data/iris.csv')
```

前提是：
你当前运行 Notebook 的位置就是项目目录，例如：

```
Desktop/Project/
```

否则程序会报错：`No such file or directory`
