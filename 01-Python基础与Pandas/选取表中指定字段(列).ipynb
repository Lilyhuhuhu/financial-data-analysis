```python
# 在表中创建新字段
import pandas as pd
data = {'Width': [4,3,4,7,8],        
'Length': [2,2,1,5,2]}
df = pd.DataFrame(data)
df
```

```python
# 求Area
df['Area'] = df['Width']*df['Length']
df
```

```python
#新建 Height
df['Height'] = [10, 10, 10, 10, 10]
df
```

```python
#更新已有字段数据
df['Height2'] = 20
df
```

```python
df['体积'] = df['Area']*df['Height']
df
```
