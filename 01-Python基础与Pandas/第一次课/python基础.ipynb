#  Python 语句整体介绍

## 1. Python 语句的概念

Python 程序是由各种语句（statement）组成的，每个语句都对应一次指令或一个逻辑步骤。语句决定了程序的执行顺序和结构，是 Python 代码最基本的构建单元。在金融数据分析、量化研究和策略开发中，我们通过不同类型的语句来实现数据加载、变量赋值、条件判断、循环处理以及策略组合等功能。

## 2. Python 语句的两大类型

Python 语句主要分为 **简单语句** 与 **复合语句** 两类，它们的主要区别在于结构形式和功能复杂度：

### 2.1 简单语句

简单语句通常在一行内完成，通过单步指令执行具体的操作。常见的简单语句包括变量赋值、表达式计算、import 导入库、return 返回值、break/continue 控制循环等。这类语句最常用于金融数据处理中，如计算收益率、提取字段、加载库与文件等。

### 2.2 复合语句

复合语句由多行代码组成，并通过缩进表示结构层级。典型的复合语句包括 if 条件判断、for/while 循环、try 异常处理、with 上下文管理，以及函数定义（def）和类定义（class）。复合语句是构建交易策略逻辑、回测结构、模型流程的核心，也是 Python 程序结构化表达的重要方式。

```python
# 类定义 class（构建对象——策略、回测框架常用）
# 定义一个简单的策略对象
class Strategy:
    def __init__(self, name):
        self.name = name
        self.position = 0

    def buy(self):
        self.position = 1
        print(self.name, ": 开仓买入")

    def sell(self):
        self.position = 0
        print(self.name, ": 清仓卖出")

# 创建对象并调用
s = Strategy("MeanReversion")
s.buy()
s.sell()
```

## 1. 数值型（Numeric Types）简介

在 Python 中，数值型用来表示各种“数”，是金融分析中最基础、最常用的数据类型之一。
常见的数值类型包括：

* **整数（`int`）**：如 100、−3、2025，常用于表示股票数量、交易笔数、天数等；
* **浮点数（`float`）**：如 15.6、-0.03、0.087，常用于表示价格、收益率、波动率等。

数值型的特点在于：

1. 可以直接参与加减乘除、幂运算等各种数学计算；
2. 适合表示连续或离散的数量、金额和比例；
3. 是构建金融指标（如收益率、夏普比率、波动率）和进行回归分析、因子计算的基础。

在实际的金融场景中，读取到的行情数据、财务数据、因子值等，通常都会以数值型的形式存储和运算。

---

## 2. 字符串（String）简介

字符串用于表示“文本信息”，在金融场景中同样非常常见，例如：

* 股票代码：如 `"600519"`、`"AAPL"`；
* 行业名称：如 `"银行"`、`"证券"`；
* 策略名称：如 `"均值回归策略"`、`"动量策略"`；
* 日期和时间：如 `"2025-01-01"`、`"2025-11-13"`。

字符串的主要特点是：

1. 用引号括起来（单引号或双引号都可以）；
2. 适合保存不能直接做数学运算，但对标识、说明、分组很重要的信息；
3. 可以进行拼接、截取、查找、替换等操作，便于构造报告、日志、文件名等。

在金融数据处理中，字符串常用来标记“这条数据属于谁、属于哪个时间、属于哪个策略或分组”，与数值型配合使用，可以更清晰地管理和展示结果。

---

## 3. 数值型与字符串在金融分析中的作用小结

* **数值型** 负责“计算”：价格、收益、风险、权重、资金规模等；
* **字符串** 负责“描述和标识”：谁的、哪天的、哪个标的、哪种策略；
* 在实际编程中，二者往往一起出现：

  * 用字符串标明“这是什么数据”；
  * 用数值型对这些数据进行“计算和分析”。

理解这两类基础数据类型的含义和使用场景，是后续学习 Python 语句、数据结构以及金融量化分析代码的起点。

# Python 常用数据结构总体介绍

Python 语言内置了多种灵活而强大的数据结构，用于存储、组织和管理不同类型的数据。在金融分析、量化研究以及一般编程场景中，选择合适的数据结构能够提升代码的可读性与运行效率。不同数据结构在有序性、可变性以及是否允许重复元素等方面存在差异，因此理解它们的特点对于有效编程至关重要。

Python 中常用的数据结构包括：

* **列表（list）**：一种有序的、可变的数据序列，支持任意数据类型的存放与动态修改，常用于存储价格序列、股票池等需要增删元素的集合。
* **元组（tuple）**：一种有序但不可变的序列，适合用于存放固定结构的数据，如配置参数或不可更改的信息组合。
* **字符串（str）**：由字符组成的不可变序列，用于表示文本信息，如股票代码、行业名称、日期等。
* **字典（dict）**：一种无序的键值对集合，适用于表示映射关系，如股票与对应价格、字段名称与其对应数值等结构化信息。
* **集合（set）**：一种无序且不包含重复元素的容器，常用于去重、集合运算（并集、交集、差集）等场景。

掌握这些基础数据结构的特点与适用场景，有助于在数据处理、逻辑控制以及金融建模中更加高效地组织和使用数据。


### 1.列表（List）
#### 🌟 列表的特点
- **有序**：元素有固定位置，可通过索引访问  
- **可变**：可随时添加、删除或修改元素  
- **能够存放不同类型的数据**（如整数、浮点数、字符串等）  
- **索引从 0 开始**

```python
# 列表（list）：有序、可变；可自由增删元素。
# 常用于：股票池、价格序列、时间序列等。
prices = [100, 102, 105]
prices.append(108)  # 添加元素b
print(prices)
```

```python
# 创建列表
my_list = []  # 创建一个空列表
my_list.append(1)  # 添加元素到列表
my_list.append('Hello')
my_list.append(3.14)
print(my_list)  # 输出：[1, 'Hello', 3.14]

my_list = [10, 20, 30]  # 创建具有初始值的列表
print(my_list)  # 输出：[10, 20, 30]
```

```python
# 索引和切片      
# 使用索引访问列表中的元素：my_list[index]    
# 使用切片获取某一范围内的元素：my_list[start:end]   

# 注意点：索引为正数时，从0开始，最后一个元素索引为len(my_list)-1；索引为负数时，从-1开始，倒数第一个元素索引为-1。   
# 切片时，左闭右开区间，包含起始索引对应的元素，不包含结束索引对应的元素。
my_list = ['a', 'b', 'c', 'd', 'e']
print(my_list[0])  # 输出：'a'
print(my_list[-1])  # 输出：'e'
print(my_list[1:3])  # 输出：['b', 'c']
```

```python
# 列表长度和查找

my_list = ['apple', 'banana', 'orange']
print(len(my_list))  
print('apple' in my_list)  
print('watermelon' in my_list)  
print(my_list.index('banana'))  
```

```python
# 修改和删除元素

my_list = ['apple', 'banana', 'orange']
my_list[1] = 'grape'  
print(my_list)  
del my_list[0]  
print(my_list)  
my_list.remove('grape')  
print(my_list)  
```

### ⭐ 2.元组的主要特点: 相比列表更加安全可靠。常用于策略固定参数、不可修改的配置项。

#### 1. 不可变（Immutable）
- 元组一旦创建，其中的元素 **不能被修改、添加或删除**。
- 由于不可变，元组通常用于存储 **固定不变的数据**。

#### 2. 有序（Ordered）
- 元组中的元素具有固定顺序，可以通过索引访问。
- 索引同样从 **0 开始**。

#### 3. 可包含不同类型的数据
- 元组中的元素可以是整数、字符串、浮点数，甚至可以包含列表、字典等复杂类型。

#### 4. 支持嵌套
- 元组可以包含其他元组，形成嵌套结构。

#### 5. 支持解包（Unpacking）
- 可以通过解包同时将多个变量赋值：

```python
strategy_config = ("MeanReversion", 20, 0.03)
print(strategy_config)
```

```python
# 字符串（str）：有序、不可变；存放文本，如股票代码、日期、行业名。
stock_code = "600519"
print(stock_code[0:3])  # 取前三位（600）
```

### 🧩 3.字典的主要特点

#### 1. 通过键（key）访问数据
- 每个值（value）都通过一个唯一的键来访问，而不是通过索引。
- 查询速度非常快，适用于快速查找的场景。

#### 2. 可变（Mutable）
- 字典创建后，可以随时 **添加、修改或删除** 键值对。

#### 3. 无序（Unordered）
- Python 3.7+ 虽然**保持插入顺序**，但逻辑意义上仍视为无序结构。
- 不通过位置来访问元素。

#### 4. 键必须唯一且不可变
- 键（key）不能重复。
- 键只能使用不可变类型，例如：
  - 字符串
  - 数字
  - 元组（不可变）

#### 5. 值可以是任意类型
- 值（value）可以是：
  - 数字、字符串
  - 列表
  - 字典
  - 任意对象

#### 6. 使用哈希表（Hash Table）实现
- 字典内部通过哈希表存储，因此可以 **快速找到某个键对应的值**。
- 查找效率非常高，平均时间复杂度为 O(1)。

```python
# 常用于：股票代码 → 价格 的映射、字段 → 值 的映射。
prices = {
    "AAPL": 175.3,
    "MSFT": 402.1,
    "TSLA": 245.2
}
print("苹果价格：", prices["AAPL"])
```

```python
# 添加和修改元素
student = {
    "name": "张三",
    "age": 20,
    "gender": "男"
}

# 添加元素
student["grade"] = "三年级"

# 修改元素
student["age"] = 21

# 删除元素
del student["gender"]

print(student)
```

```python
# 遍历字典
student = {
    "name": "张三",
    "age": 20,
    "gender": "男"
}

# 遍历键
for key in student:
    print(key)

# 遍历值
for value in student.values():
    print(value)
    
# 遍历键值对
for key, value in student.items():
    print(key, value)
```

```python
# keys()：返回字典中所有键的列表 
# values()：返回字典中所有值的列表  
# items()：返回字典中所有键值对的元组

student = {
    "name": "张三",
    "age": 20,
    "gender": "男"
}

print(student.keys())  
print(student.values()) 
print(student.items())
```

### ⭐ 4.集合的主要特点

#### 1. 元素唯一（No Duplicates）
- 集合中 **不会出现重复元素**。
- 插入重复元素会自动被忽略。

#### 2. 无序（Unordered）
- 集合中的元素没有固定顺序，也不支持索引访问。
- 不能通过下标访问元素。

#### 3. 可变（Mutable）
- 集合本身可以动态添加、删除元素。
- 但集合中的元素必须是 **不可变类型**（如数字、字符串、元组）。

#### 4. 支持集合运算
集合类型非常适合做数学集合操作，例如：
- **交集**（&）
- **并集**（|）
- **差集**（-）
- **对称差集**（^）

#### 5. 底层使用哈希表实现
- 集合的元素查找速度快，平均时间复杂度为 O(1)，类似于字典的键。


```python
# 集合（set）：无序、可变、元素唯一；自动去重。
# 常用于：股票池去重、集合运算（并集、交集等）。
raw_list = ["AAPL", "AAPL", "MSFT", "TSLA", "TSLA"]
unique_stocks = set(raw_list)

print("原股票池：", raw_list)
print("去重后的股票池：", unique_stocks)
```

## 算术运算符（Arithmetic Operators）

算术运算符用于执行各种数学计算，如加法、减法、乘法、除法等，是 Python 程序中最基础的运算工具。

---

### 📘 基本算术运算符

| 运算符 | 含义 | 示例 | 结果 |
|-------|------|------|------|
| `+` | 加法 | `3 + 2` | `5` |
| `-` | 减法 | `5 - 2` | `3` |
| `*` | 乘法 | `4 * 3` | `12` |
| `/` | 除法（返回浮点数） | `7 / 2` | `3.5` |
| `//` | 整除（向下取整） | `7 // 2` | `3` |
| `%` | 取余（求余数） | `7 % 2` | `1` |
| `**` | 幂运算 | `2 ** 3` | `8` |

```python
# 加法运算符
num1 = 10
num2 = 5
result = num1 + num2
print(result)  # 输出：15

# 减法运算符
num1 = 10
num2 = 5
result = num1 - num2
print(result)  # 输出：5

# 乘法运算符
num1 = 10
num2 = 5
result = num1 * num2
print(result)  # 输出：50

# 除法运算符
num1 = 10
num2 = 5
result = num1 / num2
print(result)  # 输出：2.0

# 整除运算符
num1 = 10
num2 = 3
result = num1 // num2
print(result)  # 输出：3

# 取余运算符
num1 = 10
num2 = 3
result = num1 % num2
print(result)  # 输出：1

# 幂运算符
num1 = 2
num2 = 3
result = num1 ** num2
print(result)  # 输出：8
```
