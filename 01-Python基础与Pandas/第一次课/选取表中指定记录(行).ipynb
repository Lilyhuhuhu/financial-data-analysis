## 选取行
|方法|功能|
|:---|:---|
|df.head(n)|前n行|
|df.tail(n)|后n行|
|df.sample(n)|随机抽n行|
|df.sample(frac)|按frac比例抽取|
|df.nlargest(n, columns)|选出columns最大的前n行|
|df.nsmallest(n, columns)|选出columns最小的前n行|
|``df[df['A']==5]``|将字段A等于5的所有的数据显示出来|
|``df.loc[indexs, columns]``|选择指定行列的数据|

```python
import pandas as pd

df = pd.read_csv('iris.csv')
df
```

```python
#字段的数据类型
df.dtypes
```

```python
#df有多少条数据
len(df)
```

```python
#前n个
df.head(10)
```

```python
#默认前5行
df.head()
```

```python
#后n个
df.tail()
```

```python
#随机抽取n个
df.sample(10)
```

```python
#随机抽1个
df.sample()
```

```python
#按照比例随机抽取
df.sample(frac=0.1)
```

```python
#按照某个字段，选取最大的前n个
df.nlargest(10, 'Sepal.Length')
```

```python
#按照某个字段，选取最小的前n个
df.nlargest(10, 'Sepal.Length')
```

```python
#逻辑条件
df['Sepal.Length']>6
```

### 逻辑选取
``df[df逻辑条件]``


```python
df[df['Sepal.Length']>6]
```

### df.loc[indexs, columns]
选取指定行列

- indexs可以使切片表达式，逻辑条件，某些情况下可以是字符串日期
- columns表示列相关的信息，一般是字符串或者字符串列表


```python
# 用 .loc 取出第 10~20 行的 Sepal.Width 列
df.loc[10:20, 'Sepal.Width']
```

```python
#选取某列，返回series
df['Species']
```

```python
type(df['Species'])
```

```python
#选取某列，返回series
df.Species
```

```python
#选取某列，返回dataframe
df[['Species']]
```

```python
type(df[['Species']])
```

```python
# 选取多列，返回dataframe
df[['Sepal.Length', 'Species']]
```

```python
# df.loc[rows, columns]
df.loc[1:10, ['Sepal.Length', 'Sepal.Width', 'Species']]
```
