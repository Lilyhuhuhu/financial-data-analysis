一、数据去重（drop_duplicates）
在数据分析中，重复数据会影响统计结果、平均值、缺失值判断甚至模型训练的准确性，因此必须在清洗环节进行去重。
1. 为什么会有重复数据？
• 数据多次抓取、拼接时产生重复行
• 数据库导出时分批合并
• 手工录入错误
2. Pandas 的去重方法：df.drop_duplicates()
语法：
df.drop_duplicates(inplace=False)
参数说明：
• inplace=False：默认不修改原数据，只返回去重结果
• inplace=True：直接修改原 DataFrame
3. 去重默认规则
• 默认以整行数据完全一致才算重复
• 保留第一条出现的记录
二、缺失值处理（dropna / fillna）<br>
缺失值（NaN）在实际数据中非常常见，它会影响：<br>
统计量（mean/median）<br>
绘图<br>
机器学习模型训练<br>
因此需要提前处理。<br>
1. 检查缺失值<br>
常用方法：
```python
df.isnull().sum()
```
2. 删除缺失值：df.dropna()<br>
```python
df.dropna(axis=0, inplace=False)
```python
参数：<br>
axis=0：删除含有缺失值的行<br>
axis=1：删除含有缺失值的列<br>
inplace=True/False：是否原地修改<br>

删除的优点：<br>
✔ 简单、不引入假数据<br>

缺点：<br>
✘ 数据量减少<br>
✘ 若缺失严重会丢掉大量样本<br>

3. 填充缺失值：df.fillna()<br>

语法：
```python
df.fillna(value, inplace=False)
```
value 可以是：<br>
固定值：如 0<br>
均值/中位数<br>

字典：为不同列填不同值<br>
例如
```python
df.fillna({'temperature': 35, 'windspeed': 10, 'event': '没有记录'})
```
填充优点：<br>
✔ 不损失样本<br>
✔ 便于后续分析<br>
4. 字典填法的理解<br>
例如：<br>
```python
df.fillna({'temperature': 35, 'windspeed': 10, 'event': '没有记录'})
```
表示：<br>
temperature 列的 NaN → 填 35<br>
windspeed 列的 NaN → 填 10<br>
event 列的 NaN → 填 “没有记录”<br>
这样每列按照自己的业务逻辑填充，更科学。<br>

# 数据去重

DataFrame数据类型有去重方法


``df.drop_duplicates(inplace)``

inplace是否对df原始数据进行去重操作，inplace的值为True或者False

````python
#非原地修改，没有对原始数据进行去重
df.drop_duplicates()
````

````python
#对原始数据去重
df.drop_duplicates(inplace=True)
df
````
