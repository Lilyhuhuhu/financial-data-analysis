### 4. 集合 set：不重复、无序的一堆元素
生活类比：<br>
• “这节课来上课的学生名单”，只关心谁来过，不关心排队顺序<br>
• “客户持有的独特产品种类”，一个产品只算一次<br>
特点：<br>
• 里面的元素 不重复<br>
• 无序，不能按下标取<br>
• 常用于：去重、看交集/并集/差集<br>
业务场景：<br>
• 统计某段时间内所有出现过的“城市”<br>
• 统计一个客户购买过的产品种类（不关心买几次）<br>
• 判断某个客户是否在黑名单集合中

```python
hobbies = ["Football", "Music", "Music", "Football", "Reading"]
unique_hobbies = set(hobbies)  # 去重
print(unique_hobbies)
```

```python
hobbies_A = {"Music", "Football", "Movies"}
hobbies_B = {"Movies", "Reading", "Travel"}

# 共同爱好（交集）
common = hobbies_A & hobbies_B     # {"Movies"}

# 所有提到的爱好（并集）
all_hobbies = hobbies_A | hobbies_B

# A 独有的爱好（差集）
only_A = hobbies_A - hobbies_B
```

```python
# 一个完整的业务小例子
# **场景：**做人力资源小系统，记录员工信息，再做一些统计。
# 一个员工用字典表示
emp1 = {
    "id": 1,
    "name": "David",
    "age": 25,
    "gender": "Male",
    "hobbies": ["Basketball", "Pingpang", "Hiking"],
    "salary": 12000.5,
    "is_manager": False,
    "leave_date": None   # 还没离职
}

# 多个员工放在列表里
employees = [
    emp1,
    {
        "id": 2,
        "name": "Mary",
        "age": 23,
        "gender": "Female",
        "hobbies": ["Reading", "Movies"],
        "salary": 13000.0,
        "is_manager": True,
        "leave_date": None
    }
]

# 统计平均工资
total_salary = 0
for emp in employees:
    total_salary += emp["salary"]

avg_salary = total_salary / len(employees)
print("平均工资：", avg_salary)

# 所有员工的唯一爱好集合
all_hobbies = set()
for emp in employees:
    all_hobbies |= set(emp["hobbies"])

print("所有员工的爱好种类：", all_hobbies)
```

## 四、函数
可重复使用的代码块
![](函数.png)

### 1、用生活比喻解释“函数”
(1) 函数是什么？<br>
函数 = 一台可以重复使用的小机器 / 小工具<br>
你给它一些“原料”（输入参数）<br>
它在内部按固定步骤加工（函数体代码）<br>
最后吐出一个“成品”（返回值 return）<br>
只要这台机器造好了，你就可以反复使用，不用每次从头写一大堆重复代码。<br>
### 2. 一个生活例子：打印快递单
想象你负责打印快递单，每一张上面都要写：<br>
收件人：XXX，地址：YYY，电话：ZZZ<br>
如果不用函数，你每次都要：<br>
(1) 手打“收件人：”<br>
(2) 手打名字<br>
(3) 手打“地址：”<br>
(4) 手打地址<br>
(5) 手打“电话：”<br>
(6) 手打电话<br>
很累，也容易打错。<br>
如果你有一个“打印快递单”的函数：<br>
只要告诉函数：名字、地址、电话<br>
函数就自动按同样格式帮你打印出来。<br>
### 3. 函数的三个关键点
(1) 函数名：给这台机器起名字（比如 print_express）<br>
(2) 参数：函数需要的输入（比如 name, address, phone）<br>
(3) 返回值：函数的输出结果（比如最终生成的一段文本<br>

```python
# 定义一个函数，名字叫 introduce
# 它需要两个输入参数：name（姓名）、age（年龄）
def introduce(name, age):
    # 使用 f-string 字符串格式化，把 name 和 age 填进句子里
    sentence = f"大家好，我是{name}，今年{age}岁。"
    # 使用 return 把这句话“返回”给函数的调用者
    return sentence

# 调用上面定义的函数
# 把 "小李" 传给参数 name，把 25 传给参数 age
result = introduce("小李", 25)

# 打印函数的返回结果
print(result)
```

```python
result2 = introduce("王二",30)
print(result2)
```

```python
result3 = introduce("张三",33)
print(result3)
```

```python
result4 = introduce("李四",35)
print(result4)
```

```python
# 定义一个名为 calc_total 的函数，用来计算总价
# 参数：
# - price: 商品的单价
# - quantity: 购买的数量
# - discount: 折扣（可选参数），默认值为 1.0，表示不打折
def calc_total(price, quantity, discount=1.0):
    # 先计算原始总价 = 单价 * 数量
    raw_total = price * quantity
    # 再乘上折扣系数，得到最终总价
    final_total = raw_total * discount
    # 把最终总价返回出去
    return final_total

# 调用函数：买 3 个，单价 50，不打折
total1 = calc_total(50, 3)
print("不打折的总价：", total1)

# 调用函数：买 5 个，单价 80，打 8 折（0.8）
total2 = calc_total(80, 5, discount=0.8)
print("打 8 折的总价：", total2)
```

```python
total3 = calc_total(60,10,0.5)
print("打5折的总价:",total3)
```
