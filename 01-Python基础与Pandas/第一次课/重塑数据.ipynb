# 《数据重塑》

（df.sort_values / df.rename / df.set_index / df.sort_index / df.drop）

---

## 1. 为什么要“重塑数据”？

你手里的 DataFrame，就像一屋子的衣服：

* 有的乱七八糟堆着（排序 sort）
* 有的名字实在看不下去（重命名 rename）
* 有的衣柜没有好好贴标签（设置索引 set_index）
* 还可能出现穿搭灾难品（想删掉 drop）

我们今天要做的，就是 **一次温柔的“数据收纳术”** 🧺✨。

---

# 2. sort_values：给乱糟糟的房间排个序

### 直觉理解

sort_values 就像给学生按照成绩排座位：

* 默认从矮 → 高（小 → 大）
* ascending=False 就是从高 → 矮（大 → 小）

### 示例

```python
df3.sort_values('mpg')
```

效果就像：

> “同学们！按油耗从低到高站好！”

如果你想反过来：

```python
df3.sort_values('mpg', ascending=False)
```

> “油耗最高的站前排！”

---

## 3. rename：给列换个更好听、看得懂的名字

### 直觉理解

有些列名像“horsepower”，学生会问你“马力是哪一列？”
你就可以给它改成更贴心的中文名。

### 示例

```python
df.rename(columns={'horsepower': '马力'})
```

> “来！从今天起你叫‘马力’了！”😄

---

## 4. set_index：重新贴标签，让数据更好找

### 直觉理解

索引 index 就像“抽屉上的贴纸”。

* 用姓名来做 index
* 或用成绩、课程做 index

这样查找会更快、看起来也更规范。

### 示例

```python
df.set_index('成绩')
```

效果：

> 把“成绩”变成左边的标签列。

---

## 5. sort_index：按标签的顺序整理抽屉

### 直觉理解

如果 index 是乱的（比如顺序被打乱），
sort_index 就像按顺序重新整理标签。

```python
df.sort_index()
```

---

## 6. drop：温柔地丢掉不要的列/行

### 直觉理解

drop 就像“断舍离”：

* drop(列名, axis=1) → 丢掉列
* drop(行名, axis=0) → 丢掉行

### 示例（丢掉“性别”列）

```python
df.drop('性别', axis=1)
```

---

# 7. 综合情境例子（超甜版）

假设有学生成绩表：

| 姓名    | 课程 | 成绩 | 性别 |
| ----- | -- | -- | -- |
| David | 数学 | 88 | 男  |
| Mary  | 英语 | 89 | 女  |

现在我们进行几步“温柔整理”：

### ✔ 第一步：按成绩排序

```python
df.sort_values('成绩', ascending=False)
```

### ✔ 第二步：给列名换成更规范的英文

```python
df.rename(columns={'成绩': 'score'})
```

### ✔ 第三步：把姓名设为 index

```python
df.set_index('姓名')
```

### ✔ 第四步：如果 index 乱了，再排一下

```python
df.sort_index()
```

### ✔ 第五步：不要展示性别

```python
df.drop('性别', axis=1)
```

这样一套走下来，数据像做了水疗一样焕然一新✨。

|函数或方法|功能|
|:---|:---|
|df.sort_values(by)|按照某列by对df进行排序|
|df.rename(columns)|对df的字段名进行重命名|
|df.set_index(keys)|设置df的行索引|
|df.sort_index()|按照行索引对df进行排序|


## df.sort_values(by)
- 可以对某(些)变量进行排序

````python
import pandas as pd
df1 = pd.read_csv('reshape/auto-mpg1.csv')
df1
````

````python
df2 = pd.read_csv('reshape/auto-mpg2.csv')
df2
````

````python
df3 = pd.concat([df1, df2])
df3
````

````python
# 根据mpg排序
df3.sort_values('mpg')
````

````python
# 根据mpg排序，设定排序方向
df3.sort_values('mpg', ascending=False)
````

````python
help(df3.sort_values)
````

## df.rename(columns)
按照columns字典,对df中的列名进行重命名
- columns 字典，例如columns={'y':'year'} ，将df中的字段名y更改为year


````python
# 将horsepower更改为'马力'； 将cylinders更改为'气缸数'
columns = {'horsepower':'马力',
          'cylinders':'气缸数'}

df3.rename(columns=columns)
````

````python
help(df3.rename)
````

## df.sort_index()
按照index对df进行排序

````python
df3.sort_index()
````

## df.set_index(keys)
将keys设置成行索引。

````python
df = pd.DataFrame({'姓名': ['David', 'David', 'David', 'Mary', 'Mary',
                           'Mary'],
                   '课程': ['语文', '数学', '英语', '语文', '数学', '英语'],
                   '成绩': [98, 88, 79, 93, 97, 89],
                   '性别': ['男', '男', '男', '女', '女', '女']})
df
````

````python
df.set_index('成绩')
````

````python
df.set_index('成绩').sort_index(ascending=False)
````

````python
help(df.sort_index)
````

## df.drop()
删除字段

df.drop(field, axis)
- field 待删除的字段名
- axis 删除方向。1为列，0为行。或者'columns'为列，'rows'为行

````python
df
````

````python
df.drop('姓名', axis=1)
````

````python
df.drop(0, axis=0)
````

````python
df.drop(5, axis=0)
````

````python
help(df.drop)
````
