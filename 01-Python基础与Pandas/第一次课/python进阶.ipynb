# 一、控制结构

## 1. 条件语句

在 Python 中，条件语句用于基于一定条件控制程序的流程。最常用的条件语句是 if 语句。

if 语句会检查一个条件，如果条件为真（True），则执行 if 语句块中的代码。可以通过 elif 和 else 来添加更多的条件和执行路径。

### 注意点
- if 语句后面的条件必须返回布尔值（True 或 False）。
- elif（意为 “else if”）可以检查多个条件。
- else 用于所有 if 和 elif 条件都不满足时。
- 不要忘记在条件语句后加冒号（:）。
- 确保代码块内的缩进一致。
- 避免过于复杂的嵌套条件语句，这会降低代码的可读性。

```python
# 检查一个数字是否为正数
number = 5
if number > 0:
    print("这是一个正数")
```

```python
# 检查数字是正数还是负数
number = -3
if number > 0:
    print("这是一个正数")
else:
    print("这是一个负数")
```

```python
# 检查数字是正数、负数还是零
number = 0
if number > 0:
    print("这是一个正数")
elif number < 0:
    print("这是一个负数")
else:
    print("这是零")
```

## 2. while 语句

在 Python 中，while 语句用于在满足特定条件的情况下重复执行一个代码块。它是基本的循环结构之一。

while 循环会不断检查一个条件是否为真（True），只要条件为真，循环体内的代码就会被执行。一旦条件为假（False），循环会停止。

### 注意点
- while 循环的条件必须返回布尔值。
- 必须确保循环有一个明确的结束条件，否则可能会造成死循环。
- 注意修改循环条件中涉及的变量，否则可能导致无限循环。
- 在循环内部处理数据时，要小心不要修改正在迭代的序列。
- 适时使用 break 和 continue 来控制循环。

```python
# 打印从1到5的数字
i = 1
while i <= 5:
    print(i)
    i += 1
```

```python
# 使用break退出循环
i = 1
while i <= 5:
    if i == 3:
        break
    print(i)
    i += 1
```

```python
# 使用continue跳过当前迭代
i = 0
while i < 5:
    i += 1
    if i == 3:
        continue
    print(i)
```

## 3. for 语句

for 语句同样是非常常用的循环结构，用于遍历序列（如列表、元组、字符串）中的每个元素，执行代码块。

for 循环会依次取出序列中的每个元素，将其赋值给循环变量，然后执行循环体。循环会持续进行，直到序列中的元素被遍历完毕。

### 注意点
- for 循环适用于任何可迭代对象。
- 循环变量可以在循环体内被使用。
- Python 的 for 循环不需要显式的索引变量。
- 避免在 for 循环中修改正在迭代的序列，这可能会导致不可预测的行为。
- for 循环中可以使用 break 和 continue 语句来控制循环流程。
- 使用 for-else 结构时，else 块在循环正常结束后执行，但如果循环被 break 中断，则不执行。

```python
# 遍历列表
fruits = ["python", "Java", "C++"]
for fruit in fruits:
    print(fruit)
```

```python
# 使用range遍历一定范围内的数字
for i in range(5):  # 从0到4
    print(i)
```

```python
# 遍历字典的键和值
scores = {"张三": 90, "李四": 85, "王五": 88}
for name, score in scores.items():
    print(f"{name}的分数是: {score}")
```

```python
# 使用enumerate同时获取索引和元素
fruits = ["python", "Java", "C++"]
for index, fruit in enumerate(fruits):
    print(f"序号{index}: {fruit}")
```

# 二、函数

## 1. 自定义函数

Python 中，自定义函数是一段可重用的代码块，用于执行特定任务。

函数通过定义一次，可以在程序的多个地方调用，增加代码的可读性和复用性。

自定义函数通过 def 关键字定义，后面跟函数名和圆括号。圆括号内可以包含参数，函数体内包含要执行的语句。函数可以返回值，也可以没有返回值（即默认返回 None）。

### 重点
- 函数定义以 def 关键字开始，后跟函数名和圆括号。
- 函数体内的语句要进行适当的缩进。
- 可以通过 return 语句从函数返回值。
- 函数调用时，要用正确的参数。

### 注意点
- 函数命名应该具有描述性，易于理解。
- 函数应尽量保持简洁，执行单一任务。
- 避免在函数中使用全局变量，应通过参数传递数据。
- 文档字符串（docstring）对于解释函数功能非常重要。

```python
# 定义一个简单的函数
def greet(name):
    """向人打招呼"""
    print(f"你好, {name}!")

# 调用函数
greet('张三')
```

```python
# 定义一个返回两数之和的函数
def add(a, b):
    """返回两个数的和"""
    return a + b

# 调用函数并打印结果
result = add(5, 3)
print(f"结果是: {result}")
```

```python
# 定义一个函数，计算三个数的乘积
def multiply(x, y, z):
    """返回三个数的乘积"""
    return x * y * z

# 调用函数
product = multiply(2, 3, 4)
print(f"乘积是: {product}")
```

```python
# 定义一个可接受多个参数的函数
def sum_all(*numbers):
    """返回所有传入数字的总和"""
    total = 0
    for number in numbers:
        total += number
    return total

# 调用函数
total = sum_all(1, 2, 3, 4)
print(f"总和是: {total}")
```

## 2. 函数调用

函数调用是执行已定义函数的过程。

调用函数时，程序会跳到该函数的定义处，执行函数体内的代码，然后返回到原来的位置继续执行。

函数调用通过使用函数名和圆括号完成。如果函数有参数，应在圆括号内提供实际参数值。当函数被调用时，控制权转移到该函数，直到函数执行完毕。

### 注意点
- 函数调用需要使用函数名和圆括号。
- 实际参数（调用时传入的参数）应与形式参数（函数定义中的参数）匹配。
- 函数可以返回值，也可以不返回（即返回 None）。
- 确保在调用函数前已经定义了该函数。
- 参数的数量和类型应该与函数定义中的一致。
- 如果函数有返回值，可以将其赋值给一个变量或直接用于表达式。

```python
# 定义返回两数之和的函数
def add(a, b):
    """返回两个数的和"""
    return a + b

# 调用函数并获取返回值
result = add(10, 20)
print(f"计算结果: {result}")
```

## 3. 参数

函数参数是函数定义的一部分，用于向函数传递信息或数据。参数是函数灵活性的关键，使得函数能以不同的方式执行操作。

函数在定义时可以指定一个或多个参数，这些参数在函数调用时接收外部传递的值。根据参数的类型和使用方式，Python 中的参数可以分为几类：必需参数、关键字参数、默认参数、可变参数。

### 重点
- 必需参数：在函数调用时必须传递的参数。
- 关键字参数：通过关键字来指定参数值，可以不按顺序传递参数。
- 默认参数：在定义函数时赋予默认值的参数。如果调用时未传递，则使用默认值。
- 可变参数：允许传入任意数量的参数，通常用于不确定参数个数的情况。

### 注意点
- 参数列表中必需参数必须放在前面。
- 默认参数值应该是不可变对象，以避免共享引用带来的问题。
- 使用关键字参数可以提高代码的可读性。
- 可变参数在处理参数数量不定的情况下非常有用。

```python
# 定义一个需要两个参数的函数
def print_info(name, age):
    print(f"名字: {name}, 年龄: {age}")

# 调用函数，传递两个参数
print_info('张三', 28)
```

## 4. 匿名函数

匿名函数（也称为 lambda 函数）是一种简洁的定义函数的方式。

通常用于定义一些简单的、一次性的函数，不需要像常规函数那样明确地定义。

匿名函数通过 lambda 关键字定义，其语法结构为：`lambda 参数列表: 表达式`。

这种函数可以接受任何数量的参数，但表达式只能有一个。

### 注意点
- lambda 函数体比常规函数体更简单。
- lambda 函数通常用于传递给高阶函数（如 map、filter、sorted 等）。
- lambda 函数不能包含多个表达式或包含命令，它们的设计目的是用于简单的情况。
- lambda 函数适用于编写简单的函数，对于复杂的逻辑应使用常规函数。
- 使用 lambda 函数时要注意代码的可读性，避免编写过于复杂的 lambda 表达式。

```python
# 定义一个简单的lambda函数
square = lambda x: x * x
print(square(5)) 
```

```python
# 对列表中的每个数字执行平方运算，并生成一个新的列表
numbers = [1, 2, 3, 4]
squared = list(map(lambda x: x**2, numbers))
print(squared)
```

```python
actions = {
    'add': lambda x, y: x + y,
    'mul': lambda x, y: x * y,
}

print(actions['add'](3, 5))  
print(actions['mul'](2, 4))  
```

# 三、异常处理

## 1. 基础异常处理

异常处理是编程中的一种机制，用于处理程序执行过程中发生的异常情况。

异常处理通过 try、except、finally 语句实现。

### 原理
- try 块：在这里编写可能会引发异常的代码。
- except 块：捕获并处理 try 块中发生的异常。
- finally 块：无论是否发生异常，这部分代码都会执行。

### 重点
- 使用异常处理来避免程序因错误而崩溃。
- 可以有多个 except 块来捕获不同类型的异常。
- finally 块通常用于清理工作，如关闭文件。

### 注意点
- 不建议捕获所有异常，应该针对可能发生的特定异常进行处理。
- 避免使用空的 except 块，这可能会隐藏真正的错误。
- 使用 finally 块来释放资源，如关闭文件，即使发生异常也要确保执行。

```python
# 捕获除零异常
try:
    result = 10 / 0
except ZeroDivisionError:
    print("除数不能为零")
```

```python
# 先捕获了IndexError异常，再用一个通用的except块捕获其他所有异常

try:
    list = [1, 2, 3]
    print(list[3])
except IndexError:
    print("索引超出列表范围")
except Exception as e:
    print("发生了一个异常:", e)
```

# 四、进阶

## 1. 数据结构

### 1. 列表（List）的进阶用法

列表是 Python 中最基础的数据结构之一，它是可变的，可以存储不同类型的对象。

### 重点用法
- 列表推导式：快速生成列表。
- 切片操作：访问列表的子集。
- 内置函数：如 append(), remove(), sort() 等。

```python
### 📌 列表（List）进阶用法示例

#### 1. 列表推导式（List Comprehension）

# 生成 1~5 的平方列表
squares = [x**2 for x in range(1, 6)]
print(squares)  # 输出：[1, 4, 9, 16, 25]
```

```python
# 切片
numbers = [10, 20, 30, 40, 50]

print(numbers[1:4])   # [20, 30, 40]
print(numbers[:3])    # [10, 20, 30]
print(numbers[-2:])   # [40, 50]
print(numbers[::2])   # [10, 30, 50] 每隔 1 个取一个
```

```python
# 内置函数
items = [3, 1, 4, 2]

items.append(5)       # 添加元素
items.remove(1)       # 删除元素 1
items.sort()          # 升序排序
print(items)          # 输出：[2, 3, 4, 5]
```

### 2. 元组（Tuple）的进阶用法

元组是不可变的列表，适用于保护数据不被修改。

### 重点用法
- 元组解包：快速分配元组中的元素。
- 命名元组：提高元组可读性。

```python
# 元组解包（Tuple Unpacking）
person = ("Alice", 25, "Engineer")
name, age, job = person

print(name)  # Alice
print(age)   # 25
print(job)   # Engineer
```

```python
# 索引
# 创建一个一维数组
array_1d = np.array([10, 20, 30, 40, 50])

# 访问第一个元素
print("第一个元素:", array_1d[0])
# 输出: 第一个元素: 10

# 访问最后一个元素
print("最后一个元素:", array_1d[-1])
# 输出: 最后一个元素: 50
```

```python
# 切片
# 切片：获取第2到第4个元素
slice_1 = array_1d[1:4]
print("第2到第4个元素:", slice_1)
# 输出: 第2到第4个元素: [20 30 40]

# 切片：获取从第2个元素开始的所有元素
slice_2 = array_1d[1:]
print("从第2个元素开始的所有元素:", slice_2)
# 输出: 从第2个元素开始的所有元素: [20 30 40 50]

# 切片：获取每隔一个元素的所有元素
slice_3 = array_1d[::2]
print("每隔一个元素的所有元素:", slice_3)
# 输出: 每隔一个元素的所有元素: [10 30 50]
```

```python
# 行切片和列切片
# 创建一个2x3的二维数组
array_2d = np.array([[1, 2, 3], [4, 5, 6]])

# 获取第一行的所有元素
row_1 = array_2d[0, :]
print("第一行的所有元素:", row_1)
# 输出: 第一行的所有元素: [1 2 3]

# 获取第二列的所有元素
col_2 = array_2d[:, 1]
print("第二列的所有元素:", col_2)
# 输出: 第二列的所有元素: [2 5]
```

```python
# 获取数组的转置
transposed_array = array_2d.T
print("原始数组:\n", array_2d)
print("转置后的数组:\n", transposed_array)
```

```python
# 将一维数组变换为二维数组
array_1d = np.arange(6)
reshaped_array = array_1d.reshape((2, 3))
print("原始一维数组:", array_1d)
print("变换后的二维数组:\n", reshaped_array)
```

```python
# 数组水平组合
# 创建两个一维数组
array1 = np.array([1, 2, 3])
array2 = np.array([4, 5, 6])

# 水平组合
h_combined = np.hstack((array1, array2))
print("水平组合的结果:", h_combined)
```

```python
# 创建两个二维数组
array1_2d = np.array([[1, 2], [3, 4]])
array2_2d = np.array([[5, 6], [7, 8]])

# 垂直组合
v_combined = np.vstack((array1_2d, array2_2d))
print("垂直组合的结果:\n", v_combined)
```

## 2. Pandas数据结构

### 2.1 Series介绍

#### 1. Series介绍
- **定义**：Series 是 Pandas 中的一种一维数据结构，可以存储任何类型的数据（整数、字符串、浮点数、Python 对象等）。它的本质是固定大小的字典，可以认为是带有标签的一维数组。
- **索引**：每个 Series 对象都有一个索引，提供了数据的快速访问方式。默认情况下，索引是从 0 开始的整数序列，但也可以自定义索引。
- **数据类型**：Series 可以包含任何数据类型，但在一个 Series 中的所有元素必须是相同类型的。

#### 2. 创建 Series
使用 `pd.Series(data, index)` 可以创建一个 Series 对象，其中 data 可以是列表、字典、数组等，index 是可选参数，用于自定义索引。

#### 3. 重点特性
- **可变性**：Series 的元素可以被修改。
- **大小不变**：虽然 Series 的元素可以被修改，但其大小（长度）是不变的。
- **数据对齐**：在进行操作时，Pandas 会根据索引对齐数据。

#### 4. 注意事项
- **性能**：对于大型数据集，Series 的操作可能比原生的 Python 数据结构慢。
- **内存使用**：Pandas 在处理大型数据时可能消耗较多内存。

```python
import pandas as pd

# 使用列表创建Series
data = [1, 2, 3, 4]
series = pd.Series(data)
print(series)
```

```python
# 使用自定义索引
series = pd.Series(data, index=['a', 'b', 'c', 'd'])
print(series)
```

### pivot() 和 pivot_table()

#### 介绍与原理
pivot() 和 pivot_table() 函数用于在 DataFrame 中进行数据透视，将长格式的数据变为宽格式。

#### 使用说明
- pivot() 适用于处理简单的透视操作。
- pivot_table() 适用于处理复杂的透视操作，可以进行聚合计算。

```python
# 创建DataFrame
data = {
    'Date': ['2025-01-01', '2025-01-02', '2025-01-03', '2025-01-04'],
    'City': ['A', 'B', 'A', 'B'],
    'Value': [10, 20, 15, 25]
}

df = pd.DataFrame(data)
print(df)

# 使用pivot进行数据透视
pivot_df = df.pivot(index='Date', columns='City', values='Value')
print(pivot_df)

# 使用pivot_table进行数据透视，计算每个城市的平均值
pivot_table_df = df.pivot_table(index='Date', columns='City', values='Value', aggfunc='mean')
print(pivot_table_df)

# 简而言之，pivot是把数据“展开”成表格，pivot_table是“展开 +计算”
```

### stack() 和 unstack()

#### 介绍与原理
stack() 和 unstack() 函数用于在多层索引之间进行数据重塑。

#### 使用说明
- stack() 将列标签旋转为行标签，生成一个 Series。
- unstack() 将行标签旋转为列标签，生成一个 DataFrame。


```python
# 使用stack将列标签旋转为行标签
stacked_df = pivot_table_df.stack()
print(stacked_df)

# 使用unstack将行标签旋转为列标签
unstacked_df = stacked_df.unstack()
print(unstacked_df)
```

## 热力图的创建

热力图（Heatmap）是一种用颜色编码的图表，通过颜色的深浅来展示矩阵或二维数据集中的值。它常用于可视化矩阵的密度、相关性和分布。

### 1 简介
热力图是一种直观的数据可视化方式，通过颜色变化来表示数据的强度。在矩阵中，每个单元格的颜色深浅表示相应位置的数值大小。热力图广泛用于展示数据的模式、趋势和关联性。

### 2 原理
热力图的原理基于颜色编码，一般采用渐变的颜色映射（Colormap）。数据值越高，颜色越深；数据值越低，颜色越浅。这种映射方式使得我们可以直观地识别矩阵中的高低值。

### 3 重点和注意事项
- **Colormap 的选择**：合适的颜色映射对于准确传达信息非常重要，Matplotlib 提供了多种 Colormap 供选择。
- **数据标准化**：在绘制热力图前，有时需要对数据进行标准化，确保各行或各列的值在相同范围内，以避免较大数值主导图像。
- **图例的添加**：添加图例能够帮助读者理解颜色与数值之间的对应关系。


```python
# 绘制热力图

# 生成示例数据：创建一个 10x10 的随机矩阵，元素是 0~1 之间的浮点数
data = np.random.random((10, 10))

# 绘制基本热力图
# imshow 用于将二维数据以图像的形式显示出来
# cmap='viridis' 指定颜色映射方案（colormap），这里使用常见的“viridis”
# interpolation='nearest' 表示不进行插值，每个像素块直接对应一个数据点
plt.imshow(data, cmap='viridis', interpolation='nearest')

# 添加颜色条（右侧的色条），用于显示颜色与数值的对应关系
plt.colorbar(label='Values')

# 添加图表标题
plt.title('Basic Heatmap')

# 显示图表（在窗口或 notebook 中展示绘制好的图像）
plt.show()
```

```python
# 相关性热力图

# 生成示例数据
data = np.corrcoef(np.random.random((10, 10)))

# 绘制相关性热力图
plt.imshow(data, cmap='coolwarm', interpolation='nearest')
plt.colorbar(label='Correlation')  # 添加颜色条

# 添加标题
plt.title('Correlation Heatmap')

# 显示图表
plt.show()
```

## 与 Seaborn 整合提升图表美观度

Seaborn 是建立在 Matplotlib 之上的统计数据可视化库，提供了更高层次的界面和更美观的默认样式。

通过与 Matplotlib 整合，可以借助 Seaborn 的功能轻松创建漂亮且具有统计意义的图表。

### 2 原理

Seaborn 使用 Matplotlib 作为基础绘图库，但在其上添加了更高层次的函数和样式。  
通过一系列优化和默认设定，使得图表更加美观，同时提供了更方便的 API 来处理统计数据。
