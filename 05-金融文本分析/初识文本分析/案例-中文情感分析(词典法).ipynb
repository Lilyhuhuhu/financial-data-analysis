# 一、什么是情感分析（Sentiment Analysis）？
一句话解释：  

情感分析就是让电脑自动判断一段文本的“情绪倾向”——是积极、消极，还是中性。  

你可以把它理解为：  

电脑在读评论、新闻、公告、文章时，尝试回答：  

• 情绪怎么样？（正面 / 负面）  

• 程度如何？（强烈 positive？轻微 negative？）  

常见应用包括：  

• 电商上判断用户评论是好评还是差评  

• 金融领域判断公司年报的语气是否乐观  

• 社交媒体分析大众对某个政策/事件的态度  

• 客服系统自动识别“愤怒客户”优先处理

# 二、情感分析怎么做？（两大主流方法）
方法 1：词典法（Dictionary-based）  

课堂最容易讲清楚的一种 ➜ 你现在 Notebook 中就是用这个。  

核心思想：  

提前准备好“带情绪的词”列表，然后统计文本中出现了多少。  

例：  

• 正向词典：优秀、增长、满意、顺利、改善…  

• 负向词典：亏损、下降、风险、困难、失败…  

步骤：  

1. 准备情感词典（积极词 + 消极词）

2. 对文本分词（例如用 Jieba）
 
3. 统计正向词数 / 负向词数

4. 得到情感分数（如：pos – neg）

词典法特点：简单、透明、可解释、容易教学。  

方法 2：机器学习 / 深度学习法  

不依赖词典，而是让模型自己学：  

• 经典算法：SVM、朴素贝叶斯、逻辑回归  

• 深度学习：LSTM、BERT、RoBERTa、GPT 等预训练模型  

需要大量标注数据，例如：  

• “这手机太棒了！” → 标签：正面  

• “糟糕透了，不会再买。” → 标签：负面  

模型学到语气、上下文关系、反讽等，比词典法更聪明。

````python
# 读取批量文件
# 从 pathlib 模块中导入 Path 类
# Path 的好处是操作路径更优雅、更跨平台（比 os.path 更现代）
from pathlib import Path

# 获取当前工作目录（cwd），并拼接上 datanew/txts，得到文本所在的目录
# Path().cwd() == 当前 Jupyter Notebook 所在的文件夹
# .joinpath('datanew','txts') 表示在当前路径下找到 datanew/txts 这个目录
txtdir = Path().cwd().joinpath('datanew', 'txts')

# 一个列表，用来保存每条记录
records = []

# 遍历 txtdir 目录下所有以 .txt 结尾的文件（模式匹配）
# txtdir.glob('*.txt') 返回一个生成器，每次产出一个 Path 文件对象
for file in txtdir.glob('*.txt'):

    # 将 Path 对象转成字符串路径，便于后续 open() 使用
    file = str(file)

    # 打开该 txt 文件，使用 gbk 编码读取全部内容，并保存到 text
    # 很多公司年报、旧文本使用 gbk，所以这里选用 gbk
    text = open(file, encoding='gbk').read()

    # file.split('/')[-1]：取出路径中的文件名，例如 "联美控股2017.txt"
    # [:-4] 去掉最后 4 个字符，即 ".txt"，只保留 "联美控股2017"
    raw = file.split('\\')[-1][:-4]

    # raw[:-4] → 除去最后四个字符，保留前面部分作为“公司名称”
    # 例如 raw = "联美控股2017"
    # raw[:-4] = "联美控股"
    name = raw[:-4]

    # raw[-4:] → 最后 4 个字符作为年份
    # raw[-4:] = "2017"
    year = raw[-4:]

    records.append({
        "name": name,
        "year": year,
        "text": text
    })

# 查看前3条
records[:3]
````

````python
# 也可以这样简化

from pathlib import Path

txtdir = Path().cwd().joinpath('datanew', 'txts')

records = []

for file in txtdir.glob('*.txt'):

    text = file.read_text(encoding='gbk')  # 直接用 Path 的读取

    stem = file.stem              # 文件名，不含后缀，如“三孚股份2017”
    name = stem[:-4]              # 前面的都是公司名
    year = stem[-4:]              # 最后四位是年份

    records.append({
        "name": name,
        "year": year,
        "text": text
    })

records[:1]
````

## 3. 定义计算情感的函数senti

````python
# 导入 jieba 用于中文分词
import jieba

# 定义情感分析函数 senti()
def senti(text):
    # 初始化正向词和负向词的计数器
    pos_num = 0
    neg_num = 0

    # 对文本进行中文分词，返回一个词语列表
    words = jieba.lcut(text)

    # 遍历分词列表中的每一个词语
    for word in words:

        # 如果该词在正向词典中，则正向计数 +1
        if word in poss:
            pos_num = pos_num + 1

        # 如果该词在负向词典中，则负向计数 +1
        if word in negs:
            neg_num = neg_num + 1

    # 将统计结果以字典的结构返回
    return {"pos": pos_num, "neg": neg_num}

# 查看某份年报的情绪得分
senti(text)
````

```python
# 假设 poss, negs 已经在练习 B 中读好了

pos_set = set(poss)

def senti(text):
    """
    输入：一段中文文本（字符串）
```

## 📝 练习 E 答案：解释情绪结果

**题目回顾：**

> 已知结果：
> `{"pos": 45, "neg": 4}`
> 问：
> 1）整体情绪偏正向 / 负向 / 中性？
> 2）这种“词典计数法”可能忽略什么？

**参考答案（文字版）：**

1. 由于正向词数量远大于负向词数量（45 vs 4），
   👉 可以认为：**该文本整体情绪明显偏正向、语气较为乐观**。
   在年报场景中，可以解读为：公司在报告期内整体表达的是“积极、成长、向好”的语气。

2. 词典计数法可能忽略的地方包括：

   * **语境问题**：
     “风险可控”“降低亏损”“避免失败”中可能包含负面词（风险、亏损、失败），
     但整体表达的是“控制、改善”，并不是真正的负面情绪。
   * **否定词问题**：
     “并不悲观”“不是失败”中含有负面词，但前面有“不是”“不”，实际语气可能偏中性或正向。
   * **反讽与隐晦表达**：
     例如：“真是优秀得让人无话可说”，可能是反话，词典很难识别。
   * **词典覆盖问题**：
     新词、行业术语、网络用语（“爆雷”“踩雷”“躺平”等）如果不在词典里，会被完全忽略。

> **词典法优点是简单透明、容易实现；缺点是“不会理解语境”，容易被否定句、反讽、隐含语气骗到。**

# 🎯 **sentiment 和 emotion de 核心差异总结**

### ✔ `Sentiment` —— 二分类情感分析（正向 / 负向）

### ✔ `Emotion` —— 多类别情绪识别（快乐 / 愤怒 / 悲伤 / 恐惧 / 惊讶 / 厌恶）

它们来自 **同一个库 cnsenti**，但功能完全不同。

* 内容越长、越正式的文本越适用

```python
result = senti.sentiment_count(text)
```

得到结果结构大概是：

```python
{'pos': 12, 'neg': 2}
```

🔍 意思是：

* 出现了 12 个正向词
* 出现了 2 个负向词
* 文本整体偏正向

👉 **这是和你自定义词典法最接近的分析方式**。
