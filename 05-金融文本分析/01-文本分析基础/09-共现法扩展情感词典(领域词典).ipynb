# 词共现矩阵
co_occurrence_matrix

共现的趋势，共现的倾向。

物以类聚，人以群分

相似的词更容易出现在一起。

```python
import cntext as ct

documents = ["I go to school every day by bus .",
             "i go to theatre every night by bus"]

matrix = ct.co_occurrence_matrix(documents, 
                        window_size=2, 
                        lang='english')
matrix
```

```python
matrix.to_csv('output/共现矩阵.csv')
```

```python
documents2 = ["编程很好玩",
             "Python是最好学的编程"]

ct.co_occurrence_matrix(documents2, 
                        window_size=2, 
                        lang='chinese')
```
