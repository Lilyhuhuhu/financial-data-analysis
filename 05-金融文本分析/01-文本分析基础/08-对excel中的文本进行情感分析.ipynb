### 1. 设计情感计算函数
以情感分析为例，主要计算正面和负面的信息，所以设置了pos和neg两个函数。

如果想根据词典，计算更多种指标，那么一个指标可定义一个函数。

````python
import jieba

#读取词典，生成正负情感词表
negs = open('data/dict/negative.txt', encoding='gbk').read().split('\n')
poss = open('data/dict/positive.txt', encoding='gbk').read().split('\n')

def pos(text):
    num = 0
    words = jieba.lcut(text)
    for word in words:
        if word in poss:
            num+=1
    return num

def neg(text):
    num = 0
    words = jieba.lcut(text)
    for word in words:
        if word in negs:
            num+=1
    return num

#text = '国际金融市场动荡，存在很大的投资风险'
#neg(text)
````
