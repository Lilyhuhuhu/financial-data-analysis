# 第一部分：什么是分词（Tokenization）

## 分词的直观理解

分词就是把一句话切成一个个有意义的“词”。  
因为计算机看文本是一串连续的字符，需要我们帮它“切开”。

示例：

原句：
```
我今天吃了一个苹果
```

分词后：
```
["我", "今天", "吃了", "一个", "苹果"]
```

## 为什么要分词？

因为后续所有文本分析步骤都依赖分词：

- 词频统计
- 文本分类
- 情感分析
- 主题模型
- 文本相似度计算
- 深度学习中的词向量构建

一句话概括：

> **分词是中文文本分析的最基础步骤。**


# 📘第二部分：英文分词 vs 中文分词

## ✔ 英文分词 —— 简单

英文天然有空格：

```
I love data science.
```

分词通常按空格、标点切开：
```
["I", "love", "data", "science"]
```

## 中文分词 —— 难点在于“没有空格”

例如：

```
我喜欢数据科学
```

需要算法判断哪些字组成一个词，如：

- 我
- 喜欢
- 数据科学

所以中文分词需要更复杂的算法，包括：

- 词典匹配
- 统计学习
- 深度学习模型（如 BERT）

# 第三部分：中文分词的三大思想（通俗解释）

## 1. 词典匹配（最常用）

也叫 “最大匹配法”。  
把句子从左到右，与词库进行比对，尽可能匹配最长的词。

示例：
```
我喜欢吃苹果手机
```

如果词库中有：
- 苹果
- 苹果手机
- 手机

则分词结果往往是：
```
["我", "喜欢", "吃", "苹果手机"]
```

---

## 2. 统计方法（基于大规模语料）

核心思想：**看哪些字更经常一起出现**。

例如：
- 绿色金融
- 数据科学
- 人工智能

在大规模文本中高频共同出现 → 算法认为它们是一个词。

---

## 3. 深度学习（更智能）

例如：
- CRF
- BiLSTM-CRF
- BERT 分词模型

效果更好，但成本更高。

---

## ✔ 综合总结

| 方法 | 优点 | 缺点 |
|------|------|-------|
| 词典匹配 | 快，部署简单 | 遇到新词会分错 |
| 统计方法 | 能发现新词 | 需要大量数据 |
| 深度学习 | 准确度高 | 训练复杂 |

###   jieba.load_userdict(txtfile)	
导入自定义词典txt文件，为防止某些目标词被分割，需要载入自定义词典。

**jieba.load_userdict(txtfile)**

- txt文件必须为 UTF-8 编码

### 关于自定义词典
不存在“一个官方、全面、通用的自定义词典”？**
因为中文文本的词语非常依赖：

* 行业（金融、医疗、科技、教育……）
* 任务（情感分析、新闻分类、研究论文分析……）
* 场景（银行业务 vs 电商客服 vs 舆情监测）
* 实际数据（新闻 vs 微博 vs 政府报告）

**一个场景的“正确词”在另一个场景里可能是错的。**

例子：

“苹果”

* 在科技新闻是 **公司**
* 在食品数据是 **水果**
* 在金融新闻可能指 **股票 Apple Inc**

如果你把“苹果”加入成“公司名”，结果会反过来影响食品类分词。

---

业界标准是：**

> **没有一个通用完整词典，只有“按场景构建的行业词典”。**

# 第五部分：正则表达式 + 停用词清洗

## 为什么要清洗？

分词后仍然包含：
- 标点符号
- 数字
- 无意义的高频词（“的”“了”“和”）
- 空格

因此要进行清洗。

---

正则表达式 = 用“规则”在文本里找东西的工具  
生活类比：  
🔍 正则表达式就像“筛子”：  
• 想筛掉“数字”？ → 用一个筛子  
• 想筛掉“标点”？ → 换一个筛子  
• 想筛出“所有中文”？ → 用一个新的筛子  
每个符号 = 筛子的规则。 
正则表达式不是死记硬背的，而是像字典一样查着用。

---

最推荐的正则表达式查询网站：
🥇 官方权威：MDN（Mozilla Developer Network）
👉 https://developer.mozilla.org/zh-CN/docs/Web/JavaScript/Guide/Regular_expressions
- 内容非常友好、解释清晰、中文、示例丰富。
- 适合学生查阅。

---

在线测试正则的网站（强烈推荐给学生）
👉 https://regex101.com/
- 可以：
• 输入正则
• 输入文本
• 立即看到匹配结果

---

10个常用的正则表达式
- 匹配中文 [\u4e00-\u9fff]
- 匹配数字\d
-         \d+ 一串数字，例如“12345”。
- 匹配字母数字（英文、数字） \w
-       \w+：所有英文字符、数字、下划线。
- 匹配空格  \s
-       \s+：连续空格
- “取反”（除了这些都不要） [^...]
- 代表“任意字符”     .
- 匹配多个      +
- 匹配 0 次或多次    *
- 字符集合 [abc]
- 匹配标点符号（清洗常用） [^\w\s\u4e00-\u9fff]

---

举例

- [^\w\s\u4e00-\u9fff]

- Step 1：找括号 []
- 说明这是“匹配一个字符”。
- Step 2：看到 ^（在括号里）
- 表示“取反”。
- Step 3：再看里面三类字符：
• \w → 英文数字下划线
• \s → 空格
• \u4e00-\u9fff → 中文
- Step 4：组合解释（学生一下就懂了）
- “不是中文、不是英文数字、不是空格的字符都删掉。”
- 也就是：删掉所有标点符号。

---

## 🧹 1. 正则表达式清洗（删除标点）

## ✔ 最终结果

你将得到只保留有意义的词的列表。

### 正则表达式
- 中文汉字的正则表达式 ``[\u4e00-\u9fff]``
- 非中文汉字的正则表达式 ``[^\u4e00-\u9fff]``
- re.findall(pattern, string)
- re.sub(oldpattern, new, string)

````python
import re

test = "dddd好吧的ddd好的good访问"

re.sub('[a-zA-Z]', '', test) #去掉字母
````

````python
test = 'dddd好的ddd好的good?*\\~访问'
re.sub('[^\u4e00-\u9fff]', '', test)  # 去掉非汉字
````

````python
test = 'dddd好的ddd好的good?*\\~访问'

words = re.findall('[\u4e00-\u9fff]+', test)  
text = ''.join(words)
text
````

````python
# 另外一个例子

# 原始句子（课堂示例）
text = '连续5年，中央经济工作会议都对生育政策做了部署，今年的措辞格外特别：“努力稳定新出生人口规模。”'

print("【步骤1：原文本】")
print(text)
````

````python
# 使用正则表达式删除所有中文、英文之外的符号
# r'[^\w\s\u4e00-\u9fff]' 的含义：
#   ^  取反
#   \w  单词字符（a-zA-Z0-9_）
#   \s  空白字符（空格、换行）
#   \u4e00-\u9fff 中文字符范围
# 所以就是：保留中文 + 英文 + 数字 + 下划线 + 空格，其他符号全部删掉。
text_clean = re.sub(r'[^\w\s\u4e00-\u9fff]', '', text)

print("【步骤2：去除标点后的文本】")
print(text_clean)
````

````python
import jieba
# 对清洗后的文本进行分词
# jieba.lcut() 会将句子切分成一个 list
words = jieba.lcut(text_clean)

print("【步骤3：分词结果】")
print(words)     # 横向显示（Python 默认）
````

关于停用词（Stopwords）
- 停用词就是：
- 在一段文本中出现频率非常高，但对“实际含义”贡献很小的词。
- 这些词不会影响文本的主题、情绪、分类结果，所以在 NLP 中通常需要删除。
- 你可以把它理解成：
- 停用词 = 语气词 + 虚词 + 功能词
- 它们不携带“信息价值”。

---
比如：  
- 我 非常 喜欢 金融 科技 ， 但是 他 不 太 喜欢 。
- 真正有意义的词：喜欢、金融、科技
- 其他的：
• 我、非常、但是、不、太、他
都是表达语气或过渡的词，对文本主题影响很小。
所以删除停用词可以：
✔ 让分词结果更干净
✔ 更容易提取关键词
✔ 提升情感分析准确度
✔ 让建模（TF-IDF、词向量）更精确
---

````python
# 停用词文件路径根据你的实际路径填写
stopwords_path = 'datanew/stopwords.txt'

# 读取停用词文件，每一行是一个停用词
with open(stopwords_path, encoding='utf-8') as f:
    stopwords = f.read().split('\n')   # 按行切分

print("【步骤4：停用词数量】", len(stopwords))
print("【停用词示例前20个】", stopwords[:20])
````

````python
# 逐个判断：如果词不在停用词列表中，就保留
filtered_words = [w for w in words if w not in stopwords]

print("【步骤5：去停用词后的结果】")
print(filtered_words)
````

````python
print("【竖排显示分词结果】")
for i, w in enumerate(filtered_words, 1):
    print(i, w)
````

````python
# 用 DataFrame 展示
import pandas as pd
pd.DataFrame(filtered_words, columns=['词语'])
````
