```python
import cntext as ct

text = '我今天得奖了，很高兴，我要将快乐分享大家。'

ct.sentiment(text=text,
             diction=ct.load_pkl_dict('DUTIR.pkl')['DUTIR'],
             lang='chinese')
```

```python
help(ct.sentiment)
```

```python
diy_diction = {'pos': ['高兴', '快乐', '分享'],
           'neg': ['难过', '悲伤'],
           'adv': ['很', '特别']}

text1 = '我今天得奖了，很高兴，我要将快乐分享大家。'
ct.sentiment(text=text1, 
             diction=diy_diction, 
             lang='chinese')
```
