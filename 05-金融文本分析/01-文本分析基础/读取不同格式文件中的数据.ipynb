# 第一部分：什么是文本分析？
文本分析（Text Analysis）指的是：  

让计算机帮我们自动读取大量文字 → 总结 → 提取信息 → 发现规律的过程。  

## 为什么要用计算机？  

• 人能看几十篇文章  

• 电脑能看几十万篇文章  

## 常见应用场景：  

• 舆情分析（评论是正面还是负面？）  

• 新闻情绪分析（新闻是乐观还是悲观？）  

• 政策文本分析（关键词趋势）  

• 作文分析（常用词、情绪、词汇量）  

一句话总结：
文本分析 = 让电脑帮我们“读文章 + 总结 + 找规律”

想分析文本，必须先把文件读进 Python。

下面是常见文件类别与对应的读取方法：

| 文件类型 | 常见场景 | 读取方法 |
|---------|----------|-----------|
| **TXT** | 评论.txt、小说.txt | `open().read()` |
| **CSV / Excel** | 有表格的内容 | `pd.read_csv()` / `pd.read_excel()` |
| **PDF** | 报告、说明书 | `pdfplumber` |
| **Word** | 访谈、文稿 | `python-docx` |

需要考虑encoding参数  

encoding = 文件的“文字编码方式”  

| 编码方式      | 特点         | 什么时候遇到？         |
| --------- | ---------- | --------------- |
| **utf-8** | 国际标准，新文件常用 | 网站下载、自己保存的文件    |
| **gbk**   | 老中文编码      | 某些老系统、政府部门导出的数据 |

如果 encoding 不对，会出现：  

• 乱码  

• UnicodeDecodeError

### 类比：读取文件就像“打开不同包装的面包”
• txt → 塑料包装袋（最简单，一拉就开）  

• csv/excel → 带有格子的盒装面包（有表格结构）  

• pdf/word → 密封包装，需要专门开箱工具  

Python 就是“开包装的工具”。

## 1.1 读取txt
**open(file, encoding).read()**

- **file**: txt文件路径
- **encoding**   文件编码方式主要是utf-8、gbk。如果不知道编码方式，可以结合chardet库

````python
# 尝试用 utf-8 打开 TXT
try:
    text = open('datanew/新闻.txt', encoding='utf-8').read()
except UnicodeDecodeError:
    # utf-8 不行，再尝试 gbk
    text = open('data/news.txt', encoding='gbk').read()

print(text[:300])   # 打印前 300 字
````

也可以用到库pdfplumber或docx
| 库名称 | 适用文件类型 | 擅长能力 | 主要用途 | 输出结果特点 |
|--------|--------------|-----------|-----------|---------------|
| **pdfplumber** | PDF | 提取“文本内容”最强 | 读取 PDF 文字、表格 | 输出纯文本；表格可转 DataFrame |
| **pdfdocx** | PDF / Word | 批量转换、解析结构 | PDF ↔ Word 转换，读取结构 | 输出结构化文本，包含段落与格式 |
| **python-docx (docx)** | Word（.docx） | 专门操作 Word | 读取/编辑 Word | 输出段落列表，可修改 Word 内容 |

````python
import pdfplumber

pdftext = ''

with pdfplumber.open('datanew/pdf.pdf') as pdf:
    for page in pdf.pages:
        pdftext += page.extract_text()

pdftext
````

````python
from docx import Document

doc = Document('datanew/word.docx')

for i, para in enumerate(doc.paragraphs):
    print(f'第 {i+1} 段：', para.text)
````
